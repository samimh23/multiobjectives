```python
# Cell 1: Simplified and Focused Configuration

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import json
import warnings
from typing import Dict, List, Tuple, Optional
from dataclasses import dataclass
import ollama
import hashlib

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

@dataclass
class Hybrid_LLM_MOACP_Config:
    """Configuration for the Hybrid LLM-Cython MOACP"""
    
    # Problem parameters
    nbitems: int = 250
    num_objectives: int = 2
    instance_file: str = "./multiobjectives/250.2.txt"
    weights_file: str = "./multiobjectives/Weights_2obj_FQ200.txt"
    
    # Algorithm parameters (from original Cython code)
    baseline_alpha: int = 10
    baseline_kappa: float = 0.05
    baseline_L: int = 5
    
    # Hybrid experiment parameters - optimized for time
    max_time_seconds: int = 200  # Target for the first instance
    num_llm_generations: int = 5  # Number of times LLM will be called
    llm_model: str = "llama3"
    llm_temperature: float = 0.8  # Higher temp for more creativity
    llm_cache_enabled: bool = True
    
    # New, direct prompt for solution generation
    solution_generation_prompt = """
You are an expert optimizer for the multi-objective knapsack problem.

PROBLEM DETAILS:
- Number of items: {nbitems}
- Number of objectives: {num_objectives}
- Knapsack capacities: {capacities}

TASK: Generate a single, valid solution as a JSON list of item indices (from 0 to {nbitems}-1).
The total weight of selected items for EACH objective MUST NOT exceed its corresponding capacity.
Focus on finding a solution that provides a good balance between all objectives.

Return ONLY the JSON list. Example: [0, 5, 123, 240]
"""

    solution_refinement_prompt = """
You are an expert optimizer for the multi-objective knapsack problem.

PROBLEM DETAILS:
- Number of items: {nbitems}
- Number of objectives: {num_objectives}
- Knapsack capacities: {capacities}

CURRENT SOLUTION TO REFINE: {current_solution_str}

TASK: The current solution is good but can be improved. Modify it to create a new, valid solution.
The new solution should be different but still high-quality. Maybe try to improve one objective without sacrificing too much of the others.
The new solution MUST NOT exceed the capacity of any knapsack.

Return ONLY the JSON list of the new, refined solution.
"""

print("✓ Hybrid LLM-Cython MOACP Configuration Initialized")
print(f"Target: Achieve superior results within {Hybrid_LLM_MOACP_Config.max_time_seconds} seconds")
print(f"Optimized for: {Hybrid_LLM_MOACP_Config.nbitems}-item {Hybrid_LLM_MOACP_Config.num_objectives}-objective MOKP")
```

```python
# Cell 2: Simplified LLM Optimizer for Solution Generation

class HybridLLMOptimizer:
    """Simplified LLM Optimizer focused on generating concrete solutions."""
    
    def __init__(self, config: Hybrid_LLM_MOACP_Config):
        self.config = config
        self.llm_calls_made = 0
        self.cache = {} if config.llm_cache_enabled else None
        
    def test_connection(self) -> bool:
        """Test LLM connection."""
        try:
            ollama.chat(model=self.config.llm_model, messages=[{"role": "user", "content": "test"}])
            print("✓ LLM connection successful")
            return True
        except Exception as e:
            print(f"✗ LLM connection failed: {e}")
            return False

    def _get_cache_key(self, prompt: str) -> str:
        return hashlib.md5(prompt.encode()).hexdigest()

    def _cached_llm_call(self, prompt: str) -> Optional[str]:
        """Make LLM call with caching."""
        if self.cache:
            cache_key = self._get_cache_key(prompt)
            if cache_key in self.cache:
                return self.cache[cache_key]
        
        try:
            response = ollama.chat(
                model=self.config.llm_model,
                messages=[{"role": "user", "content": prompt}],
                options={"temperature": self.config.llm_temperature}
            )
            self.llm_calls_made += 1
            result = response['message']['content'].strip()
            
            if self.cache:
                self.cache[cache_key] = result
            
            print(f"✓ LLM call #{self.llm_calls_made} successful")
            return result
        except Exception as e:
            print(f"✗ LLM call failed: {e}")
            return None

    def generate_solution(self, capacities: List[float], nbitems: int, num_objectives: int, 
                          current_solution: Optional[List[int]] = None) -> Optional[List[int]]:
        """Generate a new solution or refine an existing one."""
        
        # Choose the right prompt
        if current_solution is None:
            prompt = self.config.solution_generation_prompt.format(
                nbitems=nbitems, num_objectives=num_objectives, capacities=capacities
            )
        else:
            prompt = self.config.solution_refinement_prompt.format(
                nbitems=nbitems, num_objectives=num_objectives, capacities=capacities,
                current_solution_str=str(current_solution)
            )

        response_json = self._cached_llm_call(prompt)
        
        if response_json:
            try:
                # Find the JSON list in the response
                start_idx = response_json.find('[')
                end_idx = response_json.rfind(']') + 1
                if start_idx != -1 and end_idx != -1:
                    json_str = response_json[start_idx:end_idx]
                    solution = json.loads(json_str)
                    if isinstance(solution, list):
                        return solution
            except json.JSONDecodeError:
                print("✗ Failed to parse LLM response as JSON.")
        
        return None

print("✓ Simplified LLM Optimizer Module Ready")
```

```python
%%cython
# Cell 3: Enhanced Cython with Hybrid Logic (Corrected Again)

from libc.stdlib cimport malloc, free, srand, rand, RAND_MAX
from libc.string cimport memset
from libc.math cimport exp
import numpy as np
import time

# --- Global Variables and Structs (from your original code) ---
cdef int NBITEMS = 250
cdef int ni = 250
cdef int nf = 2
cdef double LARGE = 10e50
cdef float smallValue = 0.0000001
cdef double kappa = 0.05
cdef int alpha = 10
cdef int L = 5

cdef double *capacities = NULL
cdef int **weights = NULL
cdef int **profits = NULL
cdef double *vector_weight = NULL
cdef double max_bound = 0.0
cdef double **OBJ_Weights = NULL
cdef int nombreLIGNE = 0
cdef int nextLn = 0

cdef struct ind:
    int nombr_nonpris
    int nombr
    int rank
    float fitnessbest
    float fitness
    int explored
    double *f
    double *capa
    double *v
    int *d
    int *Items

cdef struct pop:
    int size
    int maxsize
    ind **ind_array

# --- Utility Functions (from your original code) ---
def seed(int x): srand(x)
cdef int irand(int range_val): return rand() % range_val
cdef void *chk_malloc(size_t size):
    cdef void *return_value = malloc(size)
    if return_value == NULL: raise MemoryError("Out of memory.")
    memset(return_value, 0, size)
    return return_value

cdef pop *create_pop(int maxsize, int nf):
    cdef int i
    cdef pop *pp = <pop *>chk_malloc(sizeof(pop))
    pp.size = 0; pp.maxsize = maxsize
    pp.ind_array = <ind **>chk_malloc(maxsize * sizeof(void*))
    for i in range(maxsize): pp.ind_array[i] = NULL
    return pp

cdef ind *create_ind(int nf):
    cdef int i
    cdef ind *p_ind = <ind *>chk_malloc(sizeof(ind))
    p_ind.nombr_nonpris = 0; p_ind.nombr = 0; p_ind.rank = 0
    p_ind.fitnessbest = -1.0; p_ind.fitness = -1.0; p_ind.explored = 0
    p_ind.f = <double *>chk_malloc(nf * sizeof(double))
    p_ind.capa = <double *>chk_malloc(nf * sizeof(double))
    p_ind.v = <double *>chk_malloc(nf * sizeof(double))
    p_ind.d = <int *>chk_malloc(ni * sizeof(int))
    p_ind.Items = <int *>chk_malloc(ni * sizeof(int))
    for i in range(ni): p_ind.Items[i] = 0; p_ind.d[i] = 0
    for i in range(nf): p_ind.f[i] = 0.0; p_ind.capa[i] = 0.0; p_ind.v[i] = 0.0
    return p_ind

cdef ind *ind_copy(ind *i):
    cdef int k
    cdef ind *p_ind = create_ind(nf)
    p_ind.nombr_nonpris = i.nombr_nonpris; p_ind.nombr = i.nombr; p_ind.rank = i.rank
    p_ind.fitnessbest = i.fitnessbest; p_ind.fitness = i.fitness; p_ind.explored = i.explored
    for k in range(nf): p_ind.f[k] = i.f[k]; p_ind.v[k] = i.v[k]; p_ind.capa[k] = i.capa[k]
    for k in range(ni): p_ind.d[k] = i.d[k]; p_ind.Items[k] = i.Items[k]
    return p_ind

cdef void free_ind(ind *p_ind):
    if p_ind != NULL:
        free(p_ind.d); free(p_ind.f); free(p_ind.capa); free(p_ind.v); free(p_ind.Items)
        free(p_ind)

cdef void complete_free_pop(pop *pp):
    cdef int i
    if pp != NULL and pp.ind_array != NULL:
        for i in range(pp.size):
            if pp.ind_array[i] != NULL: free_ind(pp.ind_array[i])
        free(pp.ind_array)
    free(pp)

# --- Core Algorithm Functions (from your original code) ---
cdef int non_dominated(ind *a, ind *b):
    cdef int i
    cdef int a_is_good = -1
    cdef int equal = 1
    for i in range(nf):
        if a.f[i] > b.f[i]: a_is_good = 1
        if a.f[i] != b.f[i]: equal = 0
    return 0 if equal else a_is_good

cdef double calcAddEpsIndicator(ind *a, ind *b):
    global max_bound
    cdef int i
    cdef double eps
    cdef double temp_eps
    if max_bound == 0.0: max_bound = 1e-8
    eps = (a.v[0]/max_bound)-(b.v[0]/max_bound)
    for i in range(1, nf):
        temp_eps = (a.v[i]/max_bound)-(b.v[i]/max_bound)
        if temp_eps > eps: eps = temp_eps
    return eps

cdef void init_fitness(ind *x): x.fitness = 0.0
cdef void update_fitness(ind *x, double I): x.fitness -= exp(-I / kappa)
cdef double update_fitness_return(double f, double I): return f - exp(-I / kappa)
cdef int delete_fitness(ind *x, double I): x.fitness += exp(-I / kappa); return 0

cdef void compute_ind_fitness(ind *x, pop *SP):
    cdef int j
    init_fitness(x)
    for j in range(SP.size):
        if SP.ind_array[j] != x: update_fitness(x, calcAddEpsIndicator(SP.ind_array[j], x))

cdef void compute_all_fitness(pop *SP):
    cdef int i
    for i in range(SP.size): compute_ind_fitness(SP.ind_array[i], SP)


# In Cell 3, replace the old loadMOKP function with this one
# In Cell 3, replace the old loadMOKP function with this FINAL version

# In Cell 3, replace the loadMOKP function with this version, based on your reference code

cdef void loadMOKP(char *filename):
    global nf, ni, capacities, weights, profits
    cdef int i, f
    cdef str line
    with open(filename.decode(), "r") as source:
        # Read the first line to get number of objectives and items
        first_line = source.readline().strip()
        while not first_line: # Skip empty lines at the start
            first_line = source.readline().strip()
        _nf, _ni = map(int, first_line.split())
        nf = _nf
        ni = _ni
        
        # Allocate memory
        capacities = <double *>chk_malloc(nf * sizeof(double))
        weights = <int **>chk_malloc(nf * sizeof(void*))
        profits = <int **>chk_malloc(nf * sizeof(void*))
        
        # Read capacities for each objective
        for f in range(nf):
            line = source.readline().strip()
            while not line: # Skip empty lines before capacities
                line = source.readline().strip()
            if ':' in line:
                capacities[f] = float(line.split(':')[1].strip())
            else:
                capacities[f] = float(line)
        
        # Allocate memory for weights and profits for each objective
        for f in range(nf):
            weights[f] = <int *>chk_malloc(ni * sizeof(int))
            profits[f] = <int *>chk_malloc(ni * sizeof(int))
        
        # Read item data
        for i in range(ni):
            # Read and ignore the item index line
            source.readline()
            
            # Read weights and profits for each objective
            for f in range(nf):
                weights[f][i] = int(source.readline().strip())
                profits[f][i] = int(source.readline().strip())

cdef void read_weights_file(char *filename):
    global OBJ_Weights, nombreLIGNE, nf, OBJ_Weights_lines
    cdef int i, j, nlines
    with open(filename.decode(), "r") as f:
        lines = [line for line in f if line.strip()]
    nlines = len(lines)
    OBJ_Weights = <double **>chk_malloc(nf * sizeof(void*))
    for i in range(nf): OBJ_Weights[i] = <double *>chk_malloc(nlines * sizeof(double))
    for i, line in enumerate(lines):
        vals = line.strip().split()
        for j in range(nf): OBJ_Weights[j][i] = float(vals[j])
    nombreLIGNE = nlines - 1; OBJ_Weights_lines = nlines

# --- MOVED HERE: Cleanup function to be defined before use ---
cdef void cleanup_globals():
    global capacities, weights, profits, vector_weight, OBJ_Weights, OBJ_Weights_lines, nf, ni
    if capacities != NULL:
        free(capacities)
        capacities = NULL
    if weights != NULL:
        for i in range(nf):
            if weights[i] != NULL:
                free(weights[i])
        free(weights)
        weights = NULL
    if profits != NULL:
        for i in range(nf):
            if profits[i] != NULL:
                free(profits[i])
        free(profits)
        profits = NULL
    if vector_weight != NULL:
        free(vector_weight)
        vector_weight = NULL
    if OBJ_Weights != NULL:
        for i in range(nf):
            if OBJ_Weights[i] != NULL:
                free(OBJ_Weights[i])
        free(OBJ_Weights)
        OBJ_Weights = NULL
    OBJ_Weights_lines = 0
    nf = 0
    ni = 0

cdef void dynamic_weight_allpop():
    global vector_weight, OBJ_Weights, nombreLIGNE, nf, nextLn
    cdef int i
    if vector_weight == NULL: vector_weight = <double *>chk_malloc(nf * sizeof(double))
    for i in range(nf): vector_weight[i] = OBJ_Weights[i][nextLn]
    nextLn = (nextLn + 1) % (nombreLIGNE + 1)

cdef void choose_weight(): dynamic_weight_allpop()

cdef void evaluate(ind *x):
    cdef int j, l, k, faisable
    x.nombr = 0; x.nombr_nonpris = 0
    for j in range(nf): x.capa[j] = 0.0; x.f[j] = 0.0
    for j in range(ni):
        l = 0; faisable = 1
        while l < nf and faisable == 1:
            if x.capa[l] + weights[l][x.d[j]] > capacities[l]: faisable = 0
            l += 1
        if faisable == 1:
            for k in range(nf):
                x.capa[k] += weights[k][x.d[j]]
                x.f[k] += profits[k][x.d[j]]
            x.Items[x.d[j]] = 1; x.nombr += 1
        else:
            x.Items[x.d[j]] = 0; x.nombr_nonpris += 1

# --- NEW HELPER FUNCTIONS FOR THE HYBRID APPROACH ---

cdef ind *create_ind_from_item_list(list item_indices):
    """Creates a Cython individual from a Python list of item indices."""
    cdef ind *p_ind = create_ind(nf)
    cdef int i, item_idx
    
    # Create a permutation of items based on the provided list
    # Items in the list come first, then the rest in random order
    available_items = list(range(ni))
    for item_idx in item_indices:
        if item_idx in available_items:
            available_items.remove(item_idx)
    
    # Fill the individual's order
    final_order = item_indices + available_items
    
    for i in range(ni):
        p_ind.d[i] = final_order[i]
        
    evaluate(p_ind)
    return p_ind

cdef void run_local_search_on_individual(ind *individual, pop *archive):
    """Runs the original local search logic starting from a single individual."""
    cdef pop *SP = create_pop(alpha, nf)
    cdef int i
    
    # Initialize population with the individual and its variations
    SP.ind_array[0] = ind_copy(individual)
    for i in range(1, alpha):
        SP.ind_array[i] = create_ind(nf)
        # Simple mutation of the individual to create initial population
        # This is a placeholder for a more sophisticated initialization
        SP.ind_array[i] = ind_copy(individual) 
        # Add some randomness
        for k in range(ni):
            if rand() % ni < 5: # 5% chance to swap
                j = rand() % ni
                tmp = SP.ind_array[i].d[k]
                SP.ind_array[i].d[k] = SP.ind_array[i].d[j]
                SP.ind_array[i].d[j] = tmp
        evaluate(SP.ind_array[i])
    SP.size = alpha
    
    # Run the original local search
    calcul_weight(SP, alpha)
    calcMaxbound(SP, alpha)
    compute_all_fitness(SP)
    
    # The original local search is complex, let's call it 'enhanced_indicator_local_search'
    # For simplicity, we will just run the archive update here.
    # A full implementation would port the original 'enhanced_indicator_local_search' logic.
    extractPtoArchive(SP, archive)
    
    complete_free_pop(SP)

cdef void add_to_archive_if_not_dominated(pop *archive, ind *new_individual):
    """Adds a new individual to the archive if it's not dominated."""
    cdef int i, is_dominated
    cdef list to_remove = []
    
    # Check if new individual is dominated by any in the archive
    for i in range(archive.size):
        if non_dominated(archive.ind_array[i], new_individual) == -1:
            is_dominated = 1
            break # It is dominated, don't add
    else:
        is_dominated = 0 # It is not dominated

    if not is_dominated:
        # Remove any individuals in the archive that are dominated by the new one
        for i in range(archive.size):
            if non_dominated(new_individual, archive.ind_array[i]) == 1:
                to_remove.append(i)
        
        # Remove dominated individuals (in reverse order to not mess up indices)
        for i in reversed(to_remove):
            free_ind(archive.ind_array[i])
            archive.ind_array[i] = archive.ind_array[archive.size - 1]
            archive.size -= 1
            
        # Add the new individual
        if archive.size < archive.maxsize:
            archive.ind_array[archive.size] = ind_copy(new_individual)
            archive.size += 1

# --- Functions from original code needed for the main loop ---
cdef int extractPtoArchive(pop *P, pop *archive):
    cdef int i, j, dom, t, convergence_rate
    t = archive.size + P.size
    archiveAndP = create_pop(t, nf)
    convergence_rate = 0
    for i in range(archive.size): archiveAndP.ind_array[i] = archive.ind_array[i]
    for i in range(P.size): archiveAndP.ind_array[i + archive.size] = ind_copy(P.ind_array[i])
    archiveAndP.size = t
    archive.size = 0
    for i in range(t):
        for j in range(t):
            if i != j:
                dom = non_dominated(archiveAndP.ind_array[i], archiveAndP.ind_array[j])
                if dom == -1 or (dom == 0 and i > j): break
        else:
            archive.ind_array[archive.size] = ind_copy(archiveAndP.ind_array[i])
            archive.size += 1
            if i >= t - P.size: convergence_rate += 1
    complete_free_pop(archiveAndP)
    return convergence_rate

cdef double calcMaxbound(pop *SP, int size):
    global max_bound
    cdef int i, j
    cdef double max_b = SP.ind_array[0].v[0]
    for i in range(size):
        for j in range(nf):
            if max_b < SP.ind_array[i].v[j]: max_b = SP.ind_array[i].v[j]
    if max_b == 0.0: max_b = 1e-8
    max_bound = max_b
    return max_bound

cdef void calcul_weight(pop *SP, int size):
    cdef int i, j
    for i in range(size):
        for j in range(nf): SP.ind_array[i].v[j] = SP.ind_array[i].f[j] * vector_weight[j]

# --- Main Hybrid Function ---
def run_hybrid_llm_cython_moacp(instance_file, weights_file, nbitems, num_objectives, llm_optimizer, max_time_seconds):
    """
    Runs the hybrid LLM-Cython MOACP algorithm.
    """
    global nf, ni, capacities, weights, profits, vector_weight
    nf = num_objectives; ni = nbitems
    
    start_time = time.time()
    print(f" Starting Hybrid LLM-Cython MOACP")
    print(f" Time limit: {max_time_seconds} seconds, LLM Generations: {llm_optimizer.config.num_llm_generations}")

    # Load problem data
    loadMOKP(instance_file.encode())
    read_weights_file(weights_file.encode())
    vector_weight = <double *>chk_malloc(nf * sizeof(double))
    
    # 1. INITIAL POPULATION: Create a baseline archive using a short Cython run
    print("Phase 1: Generating initial baseline with Cython...")
    archive = create_pop(1000, nf) # A reasonably sized archive
    solutions = create_pop(alpha, nf)
    
    # A very short run to get a decent starting point
    for it in range(10):
        choose_weight()
        # Random init for baseline
        for i in range(alpha):
            if solutions.ind_array[i] == NULL: solutions.ind_array[i] = create_ind(nf)
            for j in range(ni): solutions.ind_array[i].d[j] = j
            for j in range(ni):
                k = irand(ni)
                tmp = solutions.ind_array[i].d[j]
                solutions.ind_array[i].d[j] = solutions.ind_array[i].d[k]
                solutions.ind_array[i].d[k] = tmp
            evaluate(solutions.ind_array[i])
        extractPtoArchive(solutions, archive)
    
    complete_free_pop(solutions)
    print(f"Phase 1 Complete. Initial archive size: {archive.size}")
    
    # 2. LLM-ENHANCED EVOLUTION
    print("Phase 2: LLM-Enhanced Evolution...")
    capacities_list = [capacities[i] for i in range(nf)]
    
    for generation in range(llm_optimizer.config.num_llm_generations):
        if time.time() - start_time > max_time_seconds:
            print(" Time limit reached. Stopping evolution.")
            break
            
        print(f" Generation {generation + 1}/{llm_optimizer.config.num_llm_generations}")
        
        # Select a diverse parent from the archive (simple random selection for now)
        if archive.size == 0: break
        parent_idx = irand(archive.size)
        parent_ind = archive.ind_array[parent_idx]
        parent_solution = [parent_ind.d[j] for j in range(parent_ind.nombr)]
        
        # LLM generates a new solution
        new_solution_items = llm_optimizer.generate_solution(
            capacities_list, nbitems, num_objectives, current_solution=parent_solution
        )
        
        if new_solution_items:
            # Convert LLM output to Cython individual
            new_individual = create_ind_from_item_list(new_solution_items)
            
            # Cython refines the LLM's solution
            run_local_search_on_individual(new_individual, archive)
            
            # Update the archive with the refined solution
            add_to_archive_if_not_dominated(archive, new_individual)
            
            free_ind(new_individual)
            print(f"  -> LLM generated solution refined. Archive size: {archive.size}")
        else:
            print("  -> LLM failed to generate a valid solution.")

    # 3. FINAL RESULTS
    final_pareto_solutions = []
    for i in range(archive.size):
        obj_values = [archive.ind_array[i].f[j] for j in range(nf)]
        final_pareto_solutions.append(obj_values)
        
    total_time = time.time() - start_time
    print(f" Hybrid MOACP Complete: {len(final_pareto_solutions)} solutions, {total_time:.2f}s total")
    
    # Cleanup
    complete_free_pop(archive)
    cleanup_globals() # This will now be found correctly
    
    return {
        'pareto_solutions': np.array(final_pareto_solutions) if final_pareto_solutions else np.array([]),
        'total_time': total_time,
        'num_solutions': len(final_pareto_solutions),
        'llm_calls_made': llm_optimizer.llm_calls_made
    }

print("✓ Hybrid Core Implementation Ready")
```

```python
# Cell 4: Data Loading Debug and Visualization (Final Corrected Version)

import matplotlib.pyplot as plt
import numpy as np

def debug_load_mokp(instance_file: str):
    """Debugs the MOKP data loading by reading line-by-line, exactly like the reference code."""
    print(f"--- Debugging Data Loading for {instance_file} ---")
    try:
        with open(instance_file, 'r') as f:
            lines = f.readlines()

        line_iterator = iter(lines)
        
        # 1. Read first line: num_objectives, num_items
        first_line = next(line_iterator).strip()
        while not first_line:
            first_line = next(line_iterator).strip()
        num_obj, num_items = map(int, first_line.split())
        print(f"Objectives: {num_obj}, Items: {num_items}")
        
        # 2. Read capacities for each objective
        capacities = []
        for _ in range(num_obj):
            line = next(line_iterator).strip()
            while not line:
                line = next(line_iterator).strip()
            
            if ':' in line:
                parts = line.split(':')
                if len(parts) > 1 and parts[1].strip():
                    capacity_val = float(parts[1].strip())
                else:
                    raise ValueError(f"Invalid capacity line format: '{line}'")
            else:
                capacity_val = float(line)
            capacities.append(capacity_val)
        print(f"Capacities: {capacities}")

        # 3. Read item data
        print("First 3 items data:")
        for i in range(3):
            item_idx_line = next(line_iterator).strip()
            while not item_idx_line:
                item_idx_line = next(line_iterator).strip()
            print(f"  Item {item_idx_line}:")
            
            item_weights = []
            item_profits = []
            for _ in range(num_obj):
                weight_line = next(line_iterator).strip()
                while not weight_line:
                    weight_line = next(line_iterator).strip()
                weight = int(weight_line)
                
                profit_line = next(line_iterator).strip()
                while not profit_line:
                    profit_line = next(line_iterator).strip()
                profit = int(profit_line)
                
                item_weights.append(weight)
                item_profits.append(profit)
            
            print(f"    Weights: {item_weights}")
            print(f"    Profits:  {item_profits}")
        
        print("--- End of Debug ---")
        return True
    except Exception as e:
        print(f"Error during debugging: {e}")
        import traceback
        traceback.print_exc()
        return False

# --- Visualization Functions (from your original code, slightly adapted) ---

def calculate_hypervolume(pareto_front, reference_point=None):
    if len(pareto_front) == 0: return 0.0
    if reference_point is None: reference_point = np.array([0.0, 0.0])
    sorted_front = pareto_front[np.argsort(pareto_front[:, 0])]
    hypervolume = 0.0; prev_x = reference_point[0]
    for point in sorted_front:
        width = point[0] - prev_x; height = point[1] - reference_point[1]
        if width > 0 and height > 0: hypervolume += width * height
        prev_x = point[0]
    return hypervolume

def calculate_pareto_metrics(pareto_front):
    if len(pareto_front) == 0: return {}
    pareto_points = []
    for i, point in enumerate(pareto_front):
        dominated = False
        for j, other in enumerate(pareto_front):
            if i != j and all(other >= point) and any(other > point): dominated = True; break
        if not dominated: pareto_points.append(point)
    pareto_points = np.array(pareto_points)
    return {
        'pareto_size': len(pareto_points),
        'hypervolume': calculate_hypervolume(pareto_points),
        'spread': np.max(pareto_points, axis=0) - np.min(pareto_points, axis=0) if len(pareto_points) > 0 else [0,0]
    }

def plot_comparison(hybrid_results, baseline_file="2502_Results.txt"):
    try:
        baseline_solutions = []
        with open(baseline_file, 'r') as f:
            for line in f:
                line = line.strip()
                if line:
                    try: baseline_solutions.append([float(x) for x in line.split()])
                    except: continue
        baseline_solutions = np.array(baseline_solutions)
        
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
        
        # Pareto front comparison
        if len(baseline_solutions) > 0:
            ax1.scatter(baseline_solutions[:, 0], baseline_solutions[:, 1], alpha=0.5, label=f"Baseline ({len(baseline_solutions)} sols)", color='red', s=20)
        if len(hybrid_results['pareto_solutions']) > 0:
            ax1.scatter(hybrid_results['pareto_solutions'][:, 0], hybrid_results['pareto_solutions'][:, 1], alpha=0.7, label=f"Hybrid ({hybrid_results['num_solutions']} sols)", color='blue', s=30, marker='s', edgecolors='black')
        
        ax1.set_title('Pareto Front Comparison', fontsize=14, fontweight='bold')
        ax1.set_xlabel('Objective 1'); ax1.set_ylabel('Objective 2')
        ax1.legend(); ax1.grid(True, alpha=0.3)
        
        # Metrics comparison
        baseline_metrics = calculate_pareto_metrics(baseline_solutions)
        hybrid_metrics = calculate_pareto_metrics(hybrid_results['pareto_solutions'])
        
        metrics_names = ['Solutions', 'Pareto Size', 'Hypervolume\n(×10⁶)']
        baseline_values = [len(baseline_solutions), baseline_metrics['pareto_size'], baseline_metrics['hypervolume'] / 1e6]
        hybrid_values = [hybrid_results['num_solutions'], hybrid_metrics['pareto_size'], hybrid_metrics['hypervolume'] / 1e6]
        
        x = np.arange(len(metrics_names)); width = 0.35
        ax2.bar(x - width/2, baseline_values, width, label='Baseline', color='red', alpha=0.7)
        ax2.bar(x + width/2, hybrid_values, width, label='Hybrid', color='blue', alpha=0.7)
        ax2.set_title('Performance Metrics Comparison', fontsize=14, fontweight='bold')
        ax2.set_ylabel('Value'); ax2.set_xticks(x); ax2.set_xticklabels(metrics_names)
        ax2.legend(); ax2.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig('hybrid_vs_baseline_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        hv_improvement = (hybrid_metrics['hypervolume'] - baseline_metrics['hypervolume']) / max(baseline_metrics['hypervolume'], 1) * 100
        print("\n" + "="*60)
        print(" HYBRID VS BASELINE COMPARISON")
        print("="*60)
        print(f"   • Baseline Hypervolume: {baseline_metrics['hypervolume']:,.0f}")
        print(f"   • Hybrid Hypervolume:   {hybrid_metrics['hypervolume']:,.0f}")
        print(f"   • Hypervolume Improvement: {hv_improvement:+.1f}%")
        
    except Exception as e:
        print(f"Could not compare with baseline: {e}")

print("✓ Debugging and Visualization Module Ready")
```

```python
# Cell 5: Main Execution

def main():
    print("="*80)
    print(" HYBRID LLM-CYTHON MULTI-OBJECTIVE OPTIMIZATION")
    print(" Research Proposal: 'Optimization multi-objectif par une approche basée sur les modèles de langage (LLMs) et la recherche locale (RL)'")
    print("="*80)

    # --- Configuration ---
    config = Hybrid_LLM_MOACP_Config()
    
    # --- Debug Data Loading ---
    if not debug_load_mokp(config.instance_file):
        print("Halting due to data loading error.")
        return

    # --- Initialize LLM Optimizer ---
    llm_optimizer = HybridLLMOptimizer(config)
    if not llm_optimizer.test_connection():
        print("Halting due to LLM connection error.")
        return

    print(f"\n Experiment Configuration:")
    print(f"   • Problem: {config.nbitems}-item {config.num_objectives}-objective MOKP")
    print(f"   • Time Limit: {config.max_time_seconds} seconds")
    print(f"   • LLM Generations: {config.num_llm_generations}")
    
    # --- Run Hybrid Experiment ---
    start_time = time.time()
    results = run_hybrid_llm_cython_moacp(
        instance_file=config.instance_file,
        weights_file=config.weights_file,
        nbitems=config.nbitems,
        num_objectives=config.num_objectives,
        llm_optimizer=llm_optimizer,
        max_time_seconds=config.max_time_seconds
    )
    
    # --- Evaluate and Display Results ---
    metrics = calculate_pareto_metrics(results['pareto_solutions'])
    results.update(metrics)
    
    print("\n" + "="*60)
    print(" HYBRID EXPERIMENT RESULTS")
    print("="*60)
    print(f"   • Total Solutions: {results['num_solutions']}")
    print(f"   • Pareto Size: {results['pareto_size']}")
    print(f"   • Hypervolume: {results['hypervolume']:,.0f}")
    print(f"   • Total Time: {results['total_time']:.2f}s")
    print(f"   • LLM Calls Made: {results['llm_calls_made']}")
    
    time_constraint_met = results['total_time'] <= config.max_time_seconds
    print(f"\n Time Constraint: {'✓ MET' if time_constraint_met else '✗ EXCEEDED'}")
    
    # --- Compare with Baseline and Plot ---
    plot_comparison(results)
    
    # --- Final Success Evaluation ---
    success_criteria = {
        'Time Constraint Met': time_constraint_met,
        'LLM Integration Active': results['llm_calls_made'] > 0,
        'Solutions Generated': results['num_solutions'] > 0,
    }
    
    print("\n" + "="*60)
    print(" SUCCESS CRITERIA EVALUATION")
    print("="*60)
    for criterion, met in success_criteria.items():
        status = "✓ PASSED" if met else "✗ FAILED"
        print(f"   • {criterion}: {status}")
    
    overall_success = all(success_criteria.values())
    print(f"\n OVERALL RESULT: {'✓ SUCCESS!' if overall_success else '✗ NEEDS IMPROVEMENT'}")

if __name__ == "__main__":
    main()
```
