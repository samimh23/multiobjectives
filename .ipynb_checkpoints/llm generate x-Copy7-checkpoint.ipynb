```python
%%cython
# ============================================================================
# HYBRID LLM-MOACP - STABLE VERSION
# ============================================================================

from libc.stdlib cimport malloc, free, srand, rand
from libc.string cimport memset
from libc.math cimport exp
import time

cdef struct ind:
    int nombr_nonpris
    int nombr
    int rank
    float fitnessbest
    float fitness
    int explored
    double *f
    double *capa
    double *v
    int *d
    int *Items

cdef struct pop:
    int size
    int maxsize
    ind **ind_array

# ============================================================================
# GLOBAL VARIABLES
# ============================================================================

cdef int NBITEMS = 250
cdef int ni = 250
cdef int L = 5
cdef double LARGE = 10e50
cdef float smallValue = 0.0000001
cdef double kappa = 0.05
cdef int alpha = 10
cdef int paretoIni = 28000

cdef int nf = 2
cdef double *capacities = NULL
cdef int **weights = NULL
cdef int **profits = NULL
cdef double *vector_weight = NULL
cdef double max_bound = 0.0
cdef double **OBJ_Weights = NULL
cdef int nombreLIGNE = 0
cdef int nextLn = 0
cdef int inv = 0
cdef int OBJ_Weights_lines = 0

# Statistics
cdef long long llm_strategy_calls = 0
cdef int current_iteration = 0

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def seed(int x):
    srand(x)

cdef int irand(int range_val):
    return rand() % range_val

cdef void *chk_malloc(size_t size):
    cdef void *return_value = malloc(size)
    if return_value == NULL:
        raise MemoryError("Out of memory.")
    memset(return_value, 0, size)
    return return_value

# ============================================================================
# POPULATION MANAGEMENT
# ============================================================================

cdef pop *create_pop(int maxsize, int nf):
    cdef int i
    cdef pop *pp = <pop *>chk_malloc(sizeof(pop))
    pp.size = 0
    pp.maxsize = maxsize
    pp.ind_array = <ind **>chk_malloc(maxsize * sizeof(void*))
    for i in range(maxsize):
        pp.ind_array[i] = NULL
    return pp

cdef ind *create_ind(int nf):
    cdef int i
    cdef ind *p_ind = <ind *>chk_malloc(sizeof(ind))
    p_ind.nombr_nonpris = 0
    p_ind.nombr = 0
    p_ind.rank = 0
    p_ind.fitnessbest = -1.0
    p_ind.fitness = -1.0
    p_ind.explored = 0
    p_ind.f = <double *>chk_malloc(nf * sizeof(double))
    p_ind.capa = <double *>chk_malloc(nf * sizeof(double))
    p_ind.v = <double *>chk_malloc(nf * sizeof(double))
    p_ind.d = <int *>chk_malloc(ni * sizeof(int))
    p_ind.Items = <int *>chk_malloc(ni * sizeof(int))
    for i in range(ni):
        p_ind.Items[i] = 0
        p_ind.d[i] = 0
    for i in range(nf):
        p_ind.f[i] = 0.0
        p_ind.capa[i] = 0.0
        p_ind.v[i] = 0.0
    return p_ind

cdef ind *ind_copy(ind *i):
    cdef ind *p_ind = create_ind(nf)
    cdef int k
    p_ind.nombr_nonpris = i.nombr_nonpris
    p_ind.nombr = i.nombr
    p_ind.rank = i.rank
    p_ind.fitnessbest = i.fitnessbest
    p_ind.fitness = i.fitness
    p_ind.explored = i.explored
    for k in range(nf):
        p_ind.f[k] = i.f[k]
        p_ind.v[k] = i.v[k]
        p_ind.capa[k] = i.capa[k]
    for k in range(ni):
        p_ind.d[k] = i.d[k]
        p_ind.Items[k] = i.Items[k]
    return p_ind

cdef void free_ind(ind *p_ind):
    if p_ind != NULL:
        free(p_ind.d)
        free(p_ind.f)
        free(p_ind.capa)
        free(p_ind.v)
        free(p_ind.Items)
        free(p_ind)

cdef void complete_free_pop(pop *pp):
    cdef int i
    if pp != NULL:
        if pp.ind_array != NULL:
            for i in range(pp.size):
                if pp.ind_array[i] != NULL:
                    free_ind(pp.ind_array[i])
                    pp.ind_array[i] = NULL
            free(pp.ind_array)
        free(pp)

cdef void cleanup_globals():
    global capacities, weights, profits, vector_weight, OBJ_Weights, OBJ_Weights_lines, nf, ni
    cdef int i
    if capacities != NULL:
        free(capacities)
        capacities = NULL
    if weights != NULL:
        for i in range(nf):
            if weights[i] != NULL:
                free(weights[i])
        free(weights)
        weights = NULL
    if profits != NULL:
        for i in range(nf):
            if profits[i] != NULL:
                free(profits[i])
        free(profits)
        profits = NULL
    if vector_weight != NULL:
        free(vector_weight)
        vector_weight = NULL
    if OBJ_Weights != NULL:
        for i in range(nf):
            if OBJ_Weights[i] != NULL:
                free(OBJ_Weights[i])
        free(OBJ_Weights)
        OBJ_Weights = NULL
    OBJ_Weights_lines = 0
    nf = 0
    ni = 0

# ============================================================================
# PARETO DOMINANCE
# ============================================================================

cdef int non_dominated(ind *p_ind_a, ind *p_ind_b):
    cdef int i
    cdef int a_is_good = -1
    cdef int equal = 1
    for i in range(nf):
        if p_ind_a.f[i] > p_ind_b.f[i]:
            a_is_good = 1
        if p_ind_a.f[i] != p_ind_b.f[i]:
            equal = 0
    if equal:
        return 0
    return a_is_good

# ============================================================================
# EPSILON INDICATOR & FITNESS
# ============================================================================

cdef double calcAddEpsIndicator(ind *p_ind_a, ind *p_ind_b):
    global max_bound
    cdef int i
    cdef double eps
    cdef double temp_eps
    if max_bound == 0.0:
        max_bound = 1e-8
    eps = (p_ind_a.v[0]/max_bound)-(p_ind_b.v[0]/max_bound)
    for i in range(1, nf):
        temp_eps = (p_ind_a.v[i]/max_bound)-(p_ind_b.v[i]/max_bound)
        if temp_eps > eps:
            eps = temp_eps
    return eps

cdef void init_fitness(ind *x):
    x.fitness = 0.0

cdef void update_fitness(ind *x, double I):
    x.fitness -= exp(-I / kappa)

cdef double update_fitness_return(double f, double I):
    return f - exp(-I / kappa)

cdef int delete_fitness(ind *x, double I):
    x.fitness += exp(-I / kappa)
    return 0

cdef void compute_ind_fitness(ind *x, pop *SP):
    cdef int j
    init_fitness(x)
    for j in range(SP.size):
        if SP.ind_array[j] != x:
            update_fitness(x, calcAddEpsIndicator(SP.ind_array[j], x))

cdef void compute_all_fitness(pop *SP):
    cdef int i
    for i in range(SP.size):
        compute_ind_fitness(SP.ind_array[i], SP)

# ============================================================================
# FILE I/O
# ============================================================================

cdef void loadMOKP(char *filename):
    global nf, ni, capacities, weights, profits
    cdef int i, f
    with open(filename.decode(), "r") as source:
        _nf, _ni = [int(x) for x in source.readline().split()]
        nf = _nf
        ni = _ni
        capacities = <double *>chk_malloc(nf * sizeof(double))
        weights = <int **>chk_malloc(nf * sizeof(void*))
        profits = <int **>chk_malloc(nf * sizeof(void*))
        for f in range(nf):
            capacities[f] = float(source.readline().strip())
            weights[f] = <int *>chk_malloc(ni * sizeof(int))
            profits[f] = <int *>chk_malloc(ni * sizeof(int))
            for i in range(ni):
                source.readline()
                weights[f][i] = int(source.readline().strip())
                profits[f][i] = int(source.readline().strip())

cdef void read_weights_file(char *filename):
    global OBJ_Weights, nombreLIGNE, nf, OBJ_Weights_lines
    cdef int i, j, nlines
    with open(filename.decode(), "r") as f:
        lines = [line for line in f if line.strip()]
    nlines = len(lines)
    OBJ_Weights = <double **>chk_malloc(nf * sizeof(void*))
    for i in range(nf):
        OBJ_Weights[i] = <double *>chk_malloc(nlines * sizeof(double))
    for i, line in enumerate(lines):
        vals = line.strip().split()
        for j in range(nf):
            OBJ_Weights[j][i] = float(vals[j])
    nombreLIGNE = nlines - 1
    OBJ_Weights_lines = nlines

cdef void dynamic_weight_allpop():
    global vector_weight, OBJ_Weights, nombreLIGNE, nf, nextLn
    cdef int i
    if vector_weight == NULL:
        vector_weight = <double *>chk_malloc(nf * sizeof(double))
    for i in range(nf):
        vector_weight[i] = OBJ_Weights[i][nextLn]
    if nextLn == nombreLIGNE:
        nextLn = 0
    else:
        nextLn += 1

cdef void choose_weight():
    dynamic_weight_allpop()

# ============================================================================
# INITIALIZATION & EVALUATION
# ============================================================================

cdef void random_init_ind(ind *x):
    cdef int j, r, tmp
    for j in range(ni):
        x.d[j] = j
    for j in range(ni):
        r = irand(ni)
        tmp = x.d[r]
        x.d[r] = x.d[j]
        x.d[j] = tmp

cdef void evaluate(ind *x):
    cdef int j, l, k, faisable
    x.nombr = 0
    x.nombr_nonpris = 0
    for j in range(nf):
        x.capa[j] = 0.0
        x.f[j] = 0.0
    for j in range(ni):
        l = 0
        faisable = 1
        while l < nf and faisable == 1:
            if x.capa[l] + weights[l][x.d[j]] > capacities[l]:
                faisable = 0
            l += 1
        if faisable == 1:
            for k in range(nf):
                x.capa[k] += weights[k][x.d[j]]
                x.f[k] += profits[k][x.d[j]]
            x.Items[x.d[j]] = 1
            x.nombr += 1
        else:
            x.Items[x.d[j]] = 0
            x.nombr_nonpris += 1

cdef void P_init_pop(pop *SP, pop *Sarchive, int alpha):
    cdef int i, x, tmp, t
    t = max(alpha, Sarchive.size)
    cdef int* shuffle = <int *>chk_malloc(t * sizeof(int))
    for i in range(t):
        shuffle[i] = i
    for i in range(t):
        x = irand(alpha)
        tmp = shuffle[i]
        shuffle[i] = shuffle[x]
        shuffle[x] = tmp
    SP.size = alpha
    if Sarchive.size > alpha:
        for i in range(alpha):
            SP.ind_array[i] = ind_copy(Sarchive.ind_array[shuffle[i]])
    else:
        for i in range(alpha):
            if shuffle[i] < Sarchive.size:
                SP.ind_array[i] = ind_copy(Sarchive.ind_array[shuffle[i]])
            else:
                SP.ind_array[i] = create_ind(nf)
                random_init_ind(SP.ind_array[i])
                evaluate(SP.ind_array[i])
    free(shuffle)

# ============================================================================
# ARCHIVE MANAGEMENT
# ============================================================================

cdef int extractPtoArchive(pop *P, pop *archive):
    cdef int i, j, dom, t, convergence_rate
    t = archive.size + P.size
    archiveAndP = create_pop(t, nf)
    convergence_rate = 0
    for i in range(archive.size):
        archiveAndP.ind_array[i] = archive.ind_array[i]
    for i in range(P.size):
        archiveAndP.ind_array[i + archive.size] = ind_copy(P.ind_array[i])
    archiveAndP.size = t
    archive.size = 0
    for i in range(t):
        for j in range(t):
            if i != j:
                dom = non_dominated(archiveAndP.ind_array[i], archiveAndP.ind_array[j])
                if dom == -1 or (dom == 0 and i > j):
                    break
        else:
            archive.ind_array[archive.size] = ind_copy(archiveAndP.ind_array[i])
            archive.size += 1
            if i >= t - P.size:
                convergence_rate += 1
    complete_free_pop(archiveAndP)
    return convergence_rate

cdef double calcMaxbound(pop *SP, int size):
    global max_bound
    cdef int i, j
    SP.size = size
    cdef double max_b = SP.ind_array[0].v[0]
    for i in range(SP.size):
        for j in range(nf):
            if max_b < SP.ind_array[i].v[j]:
                max_b = SP.ind_array[i].v[j]
    if max_b == 0.0:
        max_b = 1e-8
    max_bound = max_b
    return max_bound

cdef void calcul_weight(pop *SP, int size):
    cdef int i, j
    for i in range(SP.size):
        for j in range(nf):
            SP.ind_array[i].v[j] = SP.ind_array[i].f[j] * vector_weight[j]

# ============================================================================
# SELECTION
# ============================================================================

cdef int compute_fitness_and_select(pop *SP, ind *x, int size):
    cdef int i, worst
    cdef double worst_fit, fit_tmp
    SP.size = size
    x.fitness = 0
    compute_ind_fitness(x, SP)
    worst_fit = x.fitness
    worst = -1
    for i in range(SP.size):
        fit_tmp = update_fitness_return(SP.ind_array[i].fitness, calcAddEpsIndicator(x, SP.ind_array[i]))
        if fit_tmp > worst_fit:
            worst = i
            worst_fit = fit_tmp
    fit_tmp = x.fitness
    if worst == -1:
        return -1
    else:
        for i in range(SP.size):
            delete_fitness(SP.ind_array[i], calcAddEpsIndicator(SP.ind_array[worst], SP.ind_array[i]))
            update_fitness(SP.ind_array[i], calcAddEpsIndicator(x, SP.ind_array[i]))
        delete_fitness(x, calcAddEpsIndicator(SP.ind_array[worst], x))
        free_ind(SP.ind_array[worst])
        SP.ind_array[worst] = ind_copy(x)
        if fit_tmp - worst_fit > smallValue:
            return worst
        else:
            return -1

# ============================================================================
# STRATEGY EXECUTION FUNCTIONS
# ============================================================================

# ============================================================================
# ULTRA-FAST, SAMPLING-BASED STRATEGY EXECUTION FUNCTIONS
# ============================================================================

cdef list balanced_exploration(ind *x, list available_items, int L_val):
    cdef list scores = []
    cdef int item, k, sample_size
    cdef double eff1, eff2, mean_eff, balance, score
    cdef int feasible
    
    sample_size = 50 # Only check 50 random items instead of all 250
    
    for _ in range(sample_size):
        item = irand(ni)
        if x.Items[item] == 0:
            feasible = 1
            for k in range(nf):
                if x.capa[k] + weights[k][item] > capacities[k]:
                    feasible = 0
                    break
            if feasible:
                eff1 = profits[0][item] / (weights[0][item] + 1.0)
                eff2 = profits[1][item] / (weights[1][item] + 1.0)
                mean_eff = (eff1 + eff2) / 2.0
                balance = min(eff1, eff2) / (max(eff1, eff2) + 1e-8)
                score = mean_eff * (0.7 + 0.3 * balance)
                scores.append((score, item))
    
    scores.sort(reverse=True)
    return [item for score, item in scores[:min(L_val, len(scores))]]

cdef list objective_specialization(ind *x, list available_items, int L_val, int target_obj):
    cdef list scores = []
    cdef int item, other_obj, sample_size
    cdef double target_eff, other_eff, score
    cdef int feasible

    other_obj = 1 - target_obj
    sample_size = 50
    
    for _ in range(sample_size):
        item = irand(ni)
        if x.Items[item] == 0:
            feasible = 1
            for k in range(nf):
                if x.capa[k] + weights[k][item] > capacities[k]:
                    feasible = 0
                    break
            if feasible:
                target_eff = profits[target_obj][item] / (weights[target_obj][item] + 1.0)
                other_eff = profits[other_obj][item] / (weights[other_obj][item] + 1.0)
                score = 0.9 * target_eff + 0.1 * other_eff
                scores.append((score, item))
    
    scores.sort(reverse=True)
    return [item for score, item in scores[:L_val]]

cdef list gap_filling(ind *x, list available_items, int L_val):
    cdef list scores = []
    cdef int item, weakest_idx, sample_size
    cdef double weak_eff, total_profit, score
    cdef int feasible

    weakest_idx = 0 if x.f[0] < x.f[1] else 1
    sample_size = 50
    
    for _ in range(sample_size):
        item = irand(ni)
        if x.Items[item] == 0:
            feasible = 1
            for k in range(nf):
                if x.capa[k] + weights[k][item] > capacities[k]:
                    feasible = 0
                    break
            if feasible:
                weak_eff = profits[weakest_idx][item] / (weights[weakest_idx][item] + 1.0)
                total_profit = profits[0][item] + profits[1][item]
                score = 0.7 * weak_eff + 0.3 * total_profit / 1000.0
                scores.append((score, item))
    
    scores.sort(reverse=True)
    return [item for score, item in scores[:L_val]]

cdef list aggressive_exploitation(ind *x, list available_items, int L_val, double threshold):
    cdef list scores = []
    cdef int item, sample_size
    cdef double total_profit, total_weight, efficiency, current_avg
    cdef int feasible

    current_avg = (x.f[0] + x.f[1]) / (x.nombr + 1.0)
    sample_size = 50
    
    for _ in range(sample_size):
        item = irand(ni)
        if x.Items[item] == 0:
            feasible = 1
            for k in range(nf):
                if x.capa[k] + weights[k][item] > capacities[k]:
                    feasible = 0
                    break
            if feasible:
                total_profit = profits[0][item] + profits[1][item]
                total_weight = weights[0][item] + weights[1][item]
                efficiency = total_profit / (total_weight + 1.0)
                if efficiency > threshold * current_avg:
                    scores.append((efficiency, item))
    
    scores.sort(reverse=True)
    if scores:
        return [item for score, item in scores[:L_val]]
    else:
        return balanced_exploration(x, [], L_val)


cdef list random_perturbation(ind *x, int L_val):
    """
    A fast, sampling-based random strategy, consistent with other strategies.
    Selects up to L_val random, feasible items to add.
    """
    cdef list selected = []
    cdef int item, k, attempts
    cdef int feasible
    cdef int max_attempts = L_val * 10 # Try a few more times than needed

    attempts = 0
    while len(selected) < L_val and attempts < max_attempts:
        item = irand(ni)
        attempts += 1
        
        # Check if item is already in the solution or already selected
        if x.Items[item] == 1:
            continue
        if item in selected:
            continue
            
        # Check if the item is feasible to add
        feasible = 1
        for k in range(nf):
            if x.capa[k] + weights[k][item] > capacities[k]:
                feasible = 0
                break
        
        if feasible:
            selected.append(item)
            
    return selected
# ============================================================================
# HYBRID LOCAL SEARCH - OPTIMIZED AND STABLE VERSION (FIXED)
# ============================================================================

# In Cell 1, replace the Indicator_local_search1_hybrid with this FINAL version:

cdef void Indicator_local_search1_hybrid(pop *SP, int size, object python_strategy_provider):
    """
    A FAST and focused local search that mimics the original's search intensity.
    It only improves the population SP and does NOT manage an archive.
    """
    # --- Variable Declarations ---
    cdef ind *x
    cdef int i, j, r, sol, mino, consistant, taille, feasible, tv
    cdef int* remplace = <int *>chk_malloc(L * sizeof(int))
    cdef list selected_items
    cdef dict strategy_dict
    cdef int strategy_code
    cdef int target_obj
    cdef double threshold
    cdef int max_attempts_per_solution

    SP.size = size
    
    # Get strategy from the Python provider ONCE per local search call
    strategy_dict = python_strategy_provider.get_strategy_dict(current_iteration)
    strategy_code = strategy_dict.get('code', 1)
    target_obj = strategy_dict.get('target_obj', 0)
    threshold = strategy_dict.get('threshold', 0.9)
    
    if current_iteration % 20 == 0:
        print(f"[ITER {current_iteration}] Strategy={strategy_code}, PopSize={SP.size}")
    
    # --- Main Local Search Loop: Iterate through each solution in the population ---
    for i in range(SP.size):
        x = ind_copy(SP.ind_array[i])
        
        # --- FIX: Mimic the original's search intensity ---
        # Try to improve the solution by attempting to replace each item one by one.
        max_attempts_per_solution = x.nombr
        
        for attempt in range(max_attempts_per_solution):
            # Initialize the replacement list for this neighbor
            for l in range(L):
                remplace[l] = 0
            
            # 1. REMOVE one item (randomly)
            while True:
                mino = irand(ni)
                if x.Items[mino] == 1:
                    break
            
            x.Items[mino] = 0
            x.nombr -= 1
            x.nombr_nonpris += 1
            for r in range(nf):
                x.capa[r] -= weights[r][mino]
                x.f[r] -= profits[r][mino]
            
            # 2. SELECT items to ADD using the chosen strategy
            if strategy_code == 1:
                selected_items = balanced_exploration(x, [], L)
            elif strategy_code == 2:
                selected_items = objective_specialization(x, [], L, target_obj)
            elif strategy_code == 3:
                selected_items = gap_filling(x, [], L)
            elif strategy_code == 4:
                selected_items = aggressive_exploitation(x, [], L, threshold)
            else: # Random strategy (the true baseline)
                selected_items = random_perturbation(x, L)
            
            # 3. ADD the selected items to the solution
            taille = 0
            for item in selected_items:
                if x.Items[item] == 0:
                    consistant = 1
                    for r in range(nf):
                        if x.capa[r] + weights[r][item] > capacities[r]:
                            consistant = 0
                            break
                    if consistant == 1:
                        feasible = 1
                        for r in range(taille):
                            if item == remplace[r]:
                                feasible = 0
                                break
                        if feasible == 1:
                            remplace[taille] = item
                            taille += 1
                            x.Items[item] = 1
                            x.nombr_nonpris -= 1
                            x.nombr += 1
                            for r in range(nf):
                                x.capa[r] += weights[r][item]
                                x.f[r] += profits[r][item]
            
            # 4. EVALUATE the new neighbor solution
            for tv in range(nf):
                x.v[tv] = x.f[tv] * vector_weight[tv]
            max_bound = calcMaxbound(SP, SP.size)
            sol = compute_fitness_and_select(SP, x, SP.size)
            
            # 5. ACCEPT or REJECT the new solution
            if sol != -1:
                # ACCEPTED: The new solution 'x' has replaced a member of the population.
                # We can stop trying to improve this individual and move to the next one.
                free_ind(x) # Free the copy, as the population now has the new version
                break
            else:
                # REJECTED: Restore the solution 'x' to its previous state and try again.
                x.Items[mino] = 1
                x.nombr_nonpris -= 1
                x.nombr += 1
                for r in range(nf):
                    x.capa[r] += weights[r][mino]
                    x.f[r] += profits[r][mino]
                
                if taille >= 1:
                    for r in range(taille):
                        x.Items[remplace[r]] = 0
                        x.nombr -= 1
                        x.nombr_nonpris += 1
                        for t in range(nf):
                            x.capa[t] -= weights[t][remplace[r]]
                            x.f[t] -= profits[t][remplace[r]]
                            x.v[t] = x.f[t] * vector_weight[t]
        
        # If we exited the loop without breaking, 'x' was never accepted.
        if attempt == max_attempts_per_solution - 1:
            free_ind(x)

    # --- Clean Up ---
    free(remplace)

# ============================================================================
# MAIN RUNNER
# ============================================================================

# In Cell 1, replace the old run_hybrid_moacp with this:

def run_hybrid_moacp(instance_file, weights_file, nbitems, num_objectives, output_file, strategy_provider):
    """
    Main runner for the hybrid LLM-MOACP with a fast, decoupled local search.
    """
    global nf, ni, NBITEMS, alpha, paretoIni, L, nombreLIGNE, nextLn, inv, vector_weight
    global capacities, weights, profits, OBJ_Weights
    global current_iteration

    alpha = 10
    paretoIni = 28000
    NBL = 100
    NRUNS = 10

    for run in range(1, NRUNS+1):
        NBITEMS = nbitems
        ni = nbitems
        nf = num_objectives

        print(f"RUN {run}/{NRUNS} -- {instance_file.decode()} nbitems={ni} nf={nf} => {output_file}")

        nombreLIGNE = 0
        nextLn = 0
        inv = 0

        seed(run)
        loadMOKP(instance_file)
        read_weights_file(weights_file)

        vector_weight = <double *>chk_malloc(nf * sizeof(double))
        P = create_pop(paretoIni, nf) # This is the MAIN archive
        
        strategy_provider.reset_for_run(run)

        it = 0
        while it < NBL:
            current_iteration = it
            solutions = create_pop(alpha, nf) # This is the working population SP
            
            choose_weight()
            P_init_pop(solutions, P, alpha) # Initialize population from main archive
            calcul_weight(solutions, alpha)
            calcMaxbound(solutions, alpha)
            compute_all_fitness(solutions)
            
            # --- KEY CHANGE: Call the fast, decoupled local search ---
            Indicator_local_search1_hybrid(solutions, alpha, strategy_provider)
            
            # --- KEY CHANGE: Update the main archive AFTER the local search is finished ---
            extractPtoArchive(solutions, P)
            
            it += 1
            complete_free_pop(solutions)

        with open(output_file, "a") as fpareto:
            fpareto.write("\n")
            for i in range(P.size):
                for j in range(nf):
                    fpareto.write(f"{P.ind_array[i].f[j]:.6f} ")
                fpareto.write("\n")

        complete_free_pop(P)
        cleanup_globals()
```

```python
# ============================================================================
# CELL 2: SIMPLE STRATEGY PROVIDER
# ============================================================================

import random

class SimpleStrategyProvider:
    """
    Simple strategy provider that returns dict with strategy parameters
    """
    
    def __init__(self):
        self.current_run = 1
        self.call_count = 0
    
    def reset_for_run(self, run_number):
        """Reset for new run"""
        self.current_run = run_number
        self.call_count = 0
        print(f"\n{'='*70}")
        print(f"Strategy Provider: Starting RUN {run_number}")
        print(f"{'='*70}\n")
    
    def get_strategy_dict(self, iteration):
        """
        Return strategy dictionary with code, target_obj, and threshold
        
        Returns:
            dict with keys:
                - 'code': int (1-5)
                - 'target_obj': int (0 or 1)
                - 'threshold': float (0.85-0.95)
        """
        self.call_count += 1
        
        # Phase-based strategy selection
        if iteration < 20:
            # Early phase: Balanced exploration or gap filling
            code = random.choice([1, 3])  # 1=balanced, 3=gap_filling
        elif iteration < 60:
            # Mid phase: Diverse strategies
            code = random.choice([1, 2, 3])  # 1=balanced, 2=specialize, 3=gap_filling
        elif iteration < 90:
            # Late phase: Exploitation
            code = random.choice([3, 4])  # 3=gap_filling, 4=aggressive
        else:
            # Final phase: Pure exploitation
            code = 4  # aggressive_exploitation
        
        # Random target objective (0 or 1)
        target_obj = random.choice([0, 1])
        
        # Random threshold (0.85, 0.90, or 0.95)
        threshold = random.choice([0.85, 0.90, 0.95])
        
        return {
            'code': code,
            'target_obj': target_obj,
            'threshold': threshold
        }
    
    def get_stats(self):
        """Return statistics"""
        return {
            'total_calls': self.call_count
        }


# ============================================================================
# OPTIONAL: LLM-BASED STRATEGY PROVIDER (if you want to use LLM)
# ============================================================================

import ollama
import re

class LLMStrategyProvider:
    """
    LLM-based strategy provider (use llama3.2:3b to avoid refusals)
    """
    
    def __init__(self, model='llama3.2:3b'):
        self.model = model
        self.current_run = 1
        self.call_count = 0
        print(f"LLM Strategy Provider initialized with model: {model}")
    
    def reset_for_run(self, run_number):
        """Reset for new run"""
        self.current_run = run_number
        self.call_count = 0
        print(f"\n{'='*70}")
        print(f"LLM Strategy Provider: Starting RUN {run_number} (model: {self.model})")
        print(f"{'='*70}\n")
    
    def get_strategy_dict(self, iteration):
        """
        Use LLM to select strategy
        
        Returns:
            dict with 'code', 'target_obj', 'threshold'
        """
        self.call_count += 1
        
        # Build simple prompt
        prompt = f"""Select strategy for knapsack optimization iteration {iteration}/100.

Phase: {"Early" if iteration < 20 else "Mid" if iteration < 60 else "Late" if iteration < 90 else "Final"}

Strategies:
1 = Balanced exploration (early phase, need diversity)
2 = Objective specialization (mid phase, extreme solutions)
3 = Gap filling (always good, improve weak objective)
4 = Aggressive exploitation (late phase, high quality only)
5 = Random perturbation (if stuck)

Reply with ONE number (1-5):"""
        
        try:
            response = ollama.chat(
                model=self.model,
                messages=[{'role': 'user', 'content': prompt}],
                options={'num_predict': 10, 'temperature': 0.7}
            )
            llm_text = response['message']['content']
            
            # Extract strategy code
            match = re.search(r'[1-5]', llm_text)
            if match:
                code = int(match.group(0))
            else:
                # Fallback based on phase
                if iteration < 20:
                    code = 1
                elif iteration < 60:
                    code = 3
                else:
                    code = 4
        
        except Exception as e:
            print(f"LLM error: {e}, using fallback")
            # Fallback based on phase
            if iteration < 20:
                code = 1
            elif iteration < 60:
                code = 3
            else:
                code = 4
        
        # Random parameters
        target_obj = random.choice([0, 1])
        threshold = random.choice([0.85, 0.90, 0.95])
        
        return {
            'code': code,
            'target_obj': target_obj,
            'threshold': threshold
        }
    
    def get_stats(self):
        """Return statistics"""
        return {
            'total_calls': self.call_count,
            'model': self.model
        }
```

```python
# ============================================================================
# MAIN EXPERIMENT RUNNER - COMPLETE
# Date: 2025-11-04 19:09:29 UTC
# User: samimh23
# Target: 15-20 minutes for 10 runs
# ============================================================================

import time
import os
import numpy as np
import matplotlib.pyplot as plt
from typing import List, Tuple

# ============================================================================
# ANALYSIS FUNCTIONS
# ============================================================================

def dominates(a: List[float], b: List[float]) -> bool:
    """Check if a dominates b (maximization)"""
    return all(x >= y for x, y in zip(a, b)) and any(x > y for x, y in zip(a, b))

def get_true_pareto_front(points: List[List[float]]) -> List[List[float]]:
    """Extract non-dominated solutions"""
    pareto = []
    for i, p in enumerate(points):
        dominated = False
        for j, q in enumerate(points):
            if i != j and dominates(q, p):
                dominated = True
                break
        if not dominated:
            pareto.append(p)
    return pareto

def load_results(filename: str) -> List[List[float]]:
    """Load solutions from file"""
    points = []
    if not os.path.exists(filename):
        return points
    
    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            if line:
                try:
                    points.append([float(x) for x in line.split()])
                except:
                    pass
    return points

def calculate_hypervolume(pareto_front: List[List[float]], reference_point: List[float]) -> float:
    """Calculate 2D hypervolume (maximization)"""
    if not pareto_front or len(pareto_front[0]) != 2:
        return 0.0
    
    sorted_front = sorted(pareto_front, key=lambda p: p[0], reverse=True)
    
    hypervolume = 0.0
    prev_y = reference_point[1]
    
    for point in sorted_front:
        if point[0] <= reference_point[0] or point[1] <= reference_point[1]:
            continue
        width = point[0] - reference_point[0]
        height = point[1] - prev_y
        hypervolume += width * height
        prev_y = point[1]
    
    return hypervolume

def calculate_spacing(pareto_front: List[List[float]]) -> float:
    """Calculate spacing metric (uniformity)"""
    if len(pareto_front) < 2:
        return 0.0
    
    distances = []
    for i, sol1 in enumerate(pareto_front):
        min_dist = float('inf')
        for j, sol2 in enumerate(pareto_front):
            if i != j:
                dist = np.sqrt(sum((a - b)**2 for a, b in zip(sol1, sol2)))
                if dist < min_dist:
                    min_dist = dist
        distances.append(min_dist)
    
    mean_dist = np.mean(distances)
    spacing = np.sqrt(sum((d - mean_dist)**2 for d in distances) / len(distances))
    
    return spacing

def calculate_domination_rate(front1: List[List[float]], front2: List[List[float]]) -> float:
    """Percentage of front2 dominated by front1"""
    if not front2:
        return 0.0
    
    dominated = sum(1 for sol2 in front2 if any(dominates(sol1, sol2) for sol1 in front1))
    return 100.0 * dominated / len(front2)

# ============================================================================
# COMPARISON & VISUALIZATION
# ============================================================================

def comprehensive_comparison(baseline_file, strategic_file, label):
    """Compare baseline vs strategic results"""
    print(f"\n{'='*70}")
    print(f"COMPREHENSIVE ANALYSIS: {label}")
    print(f"{'='*70}")
    
    baseline_all = load_results(baseline_file)
    strategic_all = load_results(strategic_file)
    
    print(f"\n📊 RAW DATA:")
    print(f"  Baseline total:    {len(baseline_all)} solutions")
    print(f"  Strategic total:   {len(strategic_all)} solutions")
    
    baseline_pareto = get_true_pareto_front(baseline_all)
    strategic_pareto = get_true_pareto_front(strategic_all)
    
    print(f"\n🎯 PARETO FRONTS:")
    print(f"  Baseline Pareto:   {len(baseline_pareto)} solutions")
    print(f"  Strategic Pareto:  {len(strategic_pareto)} solutions")
    
    if baseline_pareto:
        improvement = ((len(strategic_pareto) - len(baseline_pareto)) / len(baseline_pareto) * 100)
        print(f"  Improvement:       {improvement:+.2f}%")
    
    # Calculate metrics
    all_points = baseline_pareto + strategic_pareto
    if all_points:
        ref_point = [
            min(p[0] for p in all_points) - 100,
            min(p[1] for p in all_points) - 100
        ]
        
        hv_baseline = calculate_hypervolume(baseline_pareto, ref_point)
        hv_strategic = calculate_hypervolume(strategic_pareto, ref_point)
        
        spacing_baseline = calculate_spacing(baseline_pareto)
        spacing_strategic = calculate_spacing(strategic_pareto)
        
        dom_strategic = calculate_domination_rate(strategic_pareto, baseline_pareto)
        dom_baseline = calculate_domination_rate(baseline_pareto, strategic_pareto)
        
        print(f"\n📈 QUALITY METRICS:")
        print(f"  Hypervolume:")
        print(f"    Baseline:        {hv_baseline:,.0f}")
        print(f"    Strategic:       {hv_strategic:,.0f}")
        if hv_baseline > 0:
            hv_improvement = ((hv_strategic - hv_baseline) / hv_baseline * 100)
            print(f"    Improvement:     {hv_improvement:+.2f}%")
        
        print(f"\n  Spacing (lower = better):")
        print(f"    Baseline:        {spacing_baseline:.2f}")
        print(f"    Strategic:       {spacing_strategic:.2f}")
        if spacing_baseline > 0:
            spacing_improvement = ((spacing_baseline - spacing_strategic) / spacing_baseline * 100)
            print(f"    Improvement:     {spacing_improvement:+.2f}%")
        
        print(f"\n  Domination:")
        print(f"    Strategic dominates: {dom_strategic:.1f}% of Baseline")
        print(f"    Baseline dominates:  {dom_baseline:.1f}% of Strategic")
        
        print(f"\n{'='*70}")
        if hv_strategic > hv_baseline and len(strategic_pareto) >= len(baseline_pareto):
            print("✅ RESULT: Strategic approach OUTPERFORMS baseline!")
        elif hv_strategic > hv_baseline:
            print("✅ RESULT: Strategic has BETTER quality")
        else:
            print("⚠️  RESULT: Need more analysis")
        print(f"{'='*70}\n")
        
        return baseline_pareto, strategic_pareto, {
            'hv_baseline': hv_baseline,
            'hv_strategic': hv_strategic,
            'spacing_baseline': spacing_baseline,
            'spacing_strategic': spacing_strategic,
            'dom_rate': dom_strategic
        }
    
    return baseline_pareto, strategic_pareto, {}

def plot_pareto_comparison(baseline_pareto, strategic_pareto, title, save_path=None):
    """Visualize Pareto front comparison"""
    fig, axes = plt.subplots(1, 2, figsize=(16, 7))
    
    # Plot 1: Overlay
    ax1 = axes[0]
    if baseline_pareto:
        xs_b, ys_b = zip(*baseline_pareto)
        ax1.scatter(xs_b, ys_b, 
                   label=f'Baseline ({len(baseline_pareto)} solutions)',
                   alpha=0.6, s=100, color='#1f77b4', marker='o',
                   edgecolors='black', linewidth=0.5)
    
    if strategic_pareto:
        xs_s, ys_s = zip(*strategic_pareto)
        ax1.scatter(xs_s, ys_s,
                   label=f'Strategic LLM ({len(strategic_pareto)} solutions)',
                   alpha=0.6, s=100, color='#ff7f0e', marker='^',
                   edgecolors='black', linewidth=0.5)
    
    ax1.set_xlabel('Objective 1 (Profit)', fontsize=13, fontweight='bold')
    ax1.set_ylabel('Objective 2 (Profit)', fontsize=13, fontweight='bold')
    ax1.set_title('Pareto Front Comparison', fontsize=14, fontweight='bold')
    ax1.legend(fontsize=11)
    ax1.grid(True, alpha=0.3, linestyle='--')
    
    # Plot 2: Domination
    ax2 = axes[1]
    if baseline_pareto and strategic_pareto:
        baseline_colors = []
        for sol_b in baseline_pareto:
            dominated = any(dominates(sol_s, sol_b) for sol_s in strategic_pareto)
            baseline_colors.append('#d62728' if dominated else '#2ca02c')
        
        xs_b, ys_b = zip(*baseline_pareto)
        ax2.scatter(xs_b, ys_b, c=baseline_colors,
                   alpha=0.6, s=100, marker='o',
                   edgecolors='black', linewidth=0.5)
        
        strategic_colors = []
        for sol_s in strategic_pareto:
            dominates_baseline = any(dominates(sol_s, sol_b) for sol_b in baseline_pareto)
            strategic_colors.append('#ff7f0e' if dominates_baseline else '#9467bd')
        
        xs_s, ys_s = zip(*strategic_pareto)
        ax2.scatter(xs_s, ys_s, c=strategic_colors,
                   alpha=0.6, s=120, marker='^',
                   edgecolors='black', linewidth=0.5)
    
    ax2.set_xlabel('Objective 1 (Profit)', fontsize=13, fontweight='bold')
    ax2.set_ylabel('Objective 2 (Profit)', fontsize=13, fontweight='bold')
    ax2.set_title('Domination Analysis', fontsize=14, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    
    plt.suptitle(title, fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"📊 Plot saved: {save_path}")
    
    plt.show()

def plot_strategy_usage(planner_stats, save_path=None):
    """Visualize which strategies were used"""
    strategy_dist = planner_stats.get('strategy_distribution', {})
    
    if not strategy_dist:
        print("⚠️  No strategy data to plot")
        return
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Pie chart
    strategies = list(strategy_dist.keys())
    counts = list(strategy_dist.values())
    
    # Map strategy names to readable labels
    strategy_labels = {
        'balanced_exploration': 'Balanced\nExploration',
        'objective_specialization': 'Objective\nSpecialization',
        'gap_filling': 'Gap\nFilling',
        'aggressive_exploitation': 'Aggressive\nExploitation',
        'random_perturbation': 'Random\nPerturbation'
    }
    
    labels = [strategy_labels.get(s, s) for s in strategies]
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd'][:len(strategies)]
    
    ax1.pie(counts, labels=labels, autopct='%1.1f%%', startangle=90,
            colors=colors, textprops={'fontsize': 10, 'fontweight': 'bold'})
    ax1.set_title('Strategy Distribution', fontsize=14, fontweight='bold', pad=20)
    
    # Bar chart
    ax2.barh(range(len(strategies)), counts, color=colors, alpha=0.7, edgecolor='black')
    ax2.set_yticks(range(len(strategies)))
    ax2.set_yticklabels(labels, fontsize=10)
    ax2.set_xlabel('Usage Count', fontsize=12, fontweight='bold')
    ax2.set_title('Strategy Usage Frequency', fontsize=14, fontweight='bold', pad=20)
    ax2.grid(True, alpha=0.3, axis='x', linestyle='--')
    
    # Add value labels
    for i, count in enumerate(counts):
        ax2.text(count + max(counts)*0.02, i, f'{count}', 
                va='center', fontsize=10, fontweight='bold')
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"📊 Strategy usage plot saved: {save_path}")
    
    plt.show()

# ============================================================================
# MAIN EXPERIMENT
# ============================================================================

if __name__ == "__main__":
    print(f"\n{'='*70}")
    print("STRATEGIC LLM-MOACP EXPERIMENT")
    print(f"Date: 2025-11-04 19:09:29 UTC")
    print(f"User: samimh23")
    print(f"Target: 10 runs in 15-20 minutes")
    print(f"{'='*70}\n")
    
    # Configuration
    INSTANCE_FILE = b"250.2.txt"
    WEIGHTS_FILE = b"Weights_2obj_FQ200.txt"
    NB_ITEMS = 250
    NUM_OBJECTIVES = 2
    STRATEGIC_OUTPUT = "2502_Results_STRATEGIC.txt"
    BASELINE_FILE = "2502_ResultsMD.txt"
    
    # Clean previous results
    for file in [STRATEGIC_OUTPUT, "strategy_cache.json"]:
        if os.path.exists(file):
            os.remove(file)
            print(f"🗑️  Removed: {file}")
    
    # Create strategic planner
    print("\n🚀 Initializing LLM Strategic Planner...")
    planner = LLMStrategicPlanner(
        model='llama3:8b', 
        use_cache=False
    )
    
    print(f"✅ Planner ready (model: {planner.model})\n")
    
    # Run experiment
    print(f"{'='*70}")
    print("STARTING EXPERIMENT")
    print(f"{'='*70}")
    print(f"  Instance:     {INSTANCE_FILE.decode()}")
    print(f"  Items:        {NB_ITEMS}")
    print(f"  Objectives:   {NUM_OBJECTIVES}")
    print(f"  Runs:         10")
    print(f"  Iterations:   100")
    print(f"  Output:       {STRATEGIC_OUTPUT}")
    print(f"{'='*70}\n")
    
    experiment_start = time.time()
    
    try:
        run_stats = run_strategic_moacp(
            instance_file=INSTANCE_FILE,
            weights_file=WEIGHTS_FILE,
            nbitems=NB_ITEMS,
            num_objectives=NUM_OBJECTIVES,
            output_file=STRATEGIC_OUTPUT,
            strategy_provider=planner
        )
        
        experiment_time = time.time() - experiment_start
        
        print(f"\n{'='*70}")
        print("✅ EXPERIMENT COMPLETED SUCCESSFULLY")
        print(f"{'='*70}")
        print(f"Total time:  {experiment_time:.1f}s ({experiment_time/60:.2f} min)")
        print(f"Target was:  15-20 minutes")
        
        if experiment_time < 1200:  # 20 minutes
            print(f"Status:      ✅ WITHIN TARGET")
        else:
            print(f"Status:      ⚠️  Exceeded target")
        print(f"{'='*70}\n")
        
    except Exception as e:
        print(f"\n❌ ERROR: {e}")
        import traceback
        traceback.print_exc()
        exit(1)
    
    # Get planner statistics
    planner_stats = planner.get_stats()
    
    print(f"\n{'='*70}")
    print("LLM PLANNER STATISTICS")
    print(f"{'='*70}")
    print(f"Total LLM calls:     {planner_stats['llm_calls']}")
    print(f"Cache hits:          {planner_stats['cache_hits']} ({planner_stats['hit_rate']:.1f}%)")
    print(f"Cache size:          {planner_stats['cache_size']} entries")
    print(f"\nStrategy Distribution:")
    for strategy, count in planner_stats['strategy_distribution'].items():
        pct = 100.0 * count / planner_stats['total_requests'] if planner_stats['total_requests'] > 0 else 0
        print(f"  {strategy:30s} {count:4d} ({pct:5.1f}%)")
    print(f"{'='*70}\n")
    
    # Save final cache
    planner.save_cache()
    print("💾 Strategy cache saved for future runs\n")
    
    # Analysis
    print(f"{'='*70}")
    print("COMPARATIVE ANALYSIS")
    print(f"{'='*70}\n")
    
    baseline_pareto, strategic_pareto, metrics = comprehensive_comparison(
        BASELINE_FILE,
        STRATEGIC_OUTPUT,
        f"{INSTANCE_FILE.decode()} - Strategic LLM vs Baseline"
    )
    
    # Visualization
    plot_pareto_comparison(
        baseline_pareto,
        strategic_pareto,
        title=f"Strategic LLM vs Baseline MOACP - {INSTANCE_FILE.decode()}",
        save_path="pareto_comparison_strategic.png"
    )
    
    # Strategy usage visualization
    plot_strategy_usage(planner_stats, save_path="strategy_usage.png")
    
    # Final summary
    print(f"\n{'='*70}")
    print("📋 EXECUTIVE SUMMARY")
    print(f"{'='*70}")
    print(f"Instance:                {INSTANCE_FILE.decode()}")
    print(f"Configuration:           100 iterations × 10 runs (SAME as baseline)")
    print(f"LLM Model:               {planner.model}")
    print(f"Total Runtime:           {experiment_time:.1f}s ({experiment_time/60:.1f} min)")
    print(f"Average per run:         {experiment_time/10:.1f}s ({experiment_time/600:.1f} min)")
    
    if experiment_time <= 1200:
        print(f"Target Achievement:      ✅ Within 20 min target")
    else:
        print(f"Target Achievement:      ⚠️  Exceeded 20 min target")
    
    print(f"\n📊 LLM EFFICIENCY:")
    print(f"  Total strategic calls: {planner_stats['llm_calls']}")
    print(f"  Cache efficiency:      {planner_stats['hit_rate']:.1f}%")
    print(f"  Avg calls per run:     {planner_stats['llm_calls'] / 10:.1f}")
    print(f"  Avg calls per iter:    {planner_stats['llm_calls'] / 1000:.2f}")
    
    if metrics:
        print(f"\n🎯 QUALITY COMPARISON:")
        print(f"  Pareto solutions:")
        print(f"    Baseline:            {len(baseline_pareto)}")
        print(f"    Strategic:           {len(strategic_pareto)}")
        if len(baseline_pareto) > 0:
            size_improvement = ((len(strategic_pareto) - len(baseline_pareto)) / len(baseline_pareto) * 100)
            print(f"    Improvement:         {size_improvement:+.2f}%")
        
        print(f"\n  Hypervolume:")
        print(f"    Baseline:            {metrics['hv_baseline']:,.0f}")
        print(f"    Strategic:           {metrics['hv_strategic']:,.0f}")
        if metrics['hv_baseline'] > 0:
            hv_improvement = ((metrics['hv_strategic'] - metrics['hv_baseline']) / metrics['hv_baseline'] * 100)
            print(f"    Improvement:         {hv_improvement:+.2f}%")
        
        print(f"\n  Spacing (uniformity):")
        print(f"    Baseline:            {metrics['spacing_baseline']:.4f}")
        print(f"    Strategic:           {metrics['spacing_strategic']:.4f}")
        if metrics['spacing_baseline'] > 0:
            spacing_improvement = ((metrics['spacing_baseline'] - metrics['spacing_strategic']) / metrics['spacing_baseline'] * 100)
            print(f"    Improvement:         {spacing_improvement:+.2f}%")
        
        print(f"\n  Domination Rate:")
        print(f"    Strategic dominates: {metrics['dom_rate']:.1f}% of Baseline")
    
    print(f"\n📁 GENERATED FILES:")
    print(f"  ✓ {STRATEGIC_OUTPUT}")
    print(f"  ✓ strategy_cache.json ({planner_stats['cache_size']} strategies)")
    print(f"  ✓ pareto_comparison_strategic.png")
    print(f"  ✓ strategy_usage.png")
    
    # Overall verdict
    print(f"\n{'='*70}")
    print("🏆 FINAL VERDICT")
    print(f"{'='*70}")
    
    success_criteria = {
        'runtime': experiment_time <= 1200,  # 20 minutes
        'quality': metrics.get('hv_strategic', 0) > metrics.get('hv_baseline', 0) if metrics else False,
        'pareto_size': len(strategic_pareto) >= len(baseline_pareto) if baseline_pareto else False,
        'domination': metrics.get('dom_rate', 0) > 50 if metrics else False
    }
    
    passed = sum(success_criteria.values())
    total = len(success_criteria)
    
    print(f"Success Criteria ({passed}/{total} passed):")
    print(f"  {'✅' if success_criteria['runtime'] else '❌'} Runtime ≤ 20 min:        {experiment_time/60:.1f} min")
    print(f"  {'✅' if success_criteria['quality'] else '❌'} Hypervolume improved:    {success_criteria['quality']}")
    print(f"  {'✅' if success_criteria['pareto_size'] else '❌'} Pareto size ≥ baseline: {success_criteria['pareto_size']}")
    print(f"  {'✅' if success_criteria['domination'] else '❌'} Dominates >50% baseline: {success_criteria['domination']}")
    
    print(f"\n{'='*70}")
    if passed >= 3:
        print("🎉 SUCCESS: Strategic LLM approach is EFFECTIVE!")
        print("   - Maintains computational efficiency")
        print("   - Improves solution quality")
        print("   - Suitable for research publication")
    elif passed >= 2:
        print("✅ PARTIAL SUCCESS: Strategic LLM shows promise")
        print("   - Consider tuning strategies")
        print("   - May need more runs for statistical significance")
    else:
        print("⚠️  NEEDS IMPROVEMENT: Strategic LLM underperformed")
        print("   - Review strategy selection logic")
        print("   - Consider different LLM prompting approach")
    print(f"{'='*70}\n")
    
    print("✅ ALL ANALYSES COMPLETED!\n")
```

```python
# ============================================================================
# EXTENDED ANALYSIS - Run after main experiment completes
# ============================================================================

def extended_analysis():
    """
    Comprehensive post-experiment analysis
    """
    print(f"\n{'='*70}")
    print("📊 EXTENDED ANALYSIS")
    print(f"{'='*70}\n")
    
    BASELINE_FILE = "2502_ResultsMD.txt"
    STRATEGIC_FILE = "2502_Results_STRATEGIC.txt"
    
    # Load results
    baseline_all = load_results(BASELINE_FILE)
    strategic_all = load_results(STRATEGIC_FILE)
    
    baseline_pareto = get_true_pareto_front(baseline_all)
    strategic_pareto = get_true_pareto_front(strategic_all)
    
    # 1. Unique solutions analysis
    print("1️⃣  UNIQUE SOLUTIONS ANALYSIS")
    print(f"{'='*70}")
    
    baseline_set = set(tuple(p) for p in baseline_pareto)
    strategic_set = set(tuple(p) for p in strategic_pareto)
    
    unique_to_strategic = strategic_set - baseline_set
    unique_to_baseline = baseline_set - strategic_set
    common = baseline_set & strategic_set
    
    print(f"  Common solutions:          {len(common)}")
    print(f"  Unique to Strategic:       {len(unique_to_strategic)}")
    print(f"  Unique to Baseline:        {len(unique_to_baseline)}")
    print(f"  Strategic uniqueness:      {100*len(unique_to_strategic)/len(strategic_pareto):.1f}%")
    print(f"  Baseline uniqueness:       {100*len(unique_to_baseline)/len(baseline_pareto):.1f}%\n")
    
    # 2. Objective space coverage
    print("2️⃣  OBJECTIVE SPACE COVERAGE")
    print(f"{'='*70}")
    
    if baseline_pareto and strategic_pareto:
        # Ranges
        baseline_obj1 = [p[0] for p in baseline_pareto]
        baseline_obj2 = [p[1] for p in baseline_pareto]
        strategic_obj1 = [p[0] for p in strategic_pareto]
        strategic_obj2 = [p[1] for p in strategic_pareto]
        
        print(f"  Objective 1 range:")
        print(f"    Baseline:  [{min(baseline_obj1):.0f}, {max(baseline_obj1):.0f}] (span: {max(baseline_obj1)-min(baseline_obj1):.0f})")
        print(f"    Strategic: [{min(strategic_obj1):.0f}, {max(strategic_obj1):.0f}] (span: {max(strategic_obj1)-min(strategic_obj1):.0f})")
        
        print(f"\n  Objective 2 range:")
        print(f"    Baseline:  [{min(baseline_obj2):.0f}, {max(baseline_obj2):.0f}] (span: {max(baseline_obj2)-min(baseline_obj2):.0f})")
        print(f"    Strategic: [{min(strategic_obj2):.0f}, {max(strategic_obj2):.0f}] (span: {max(strategic_obj2)-min(strategic_obj2):.0f})\n")
    
    # 3. Extreme solutions
    print("3️⃣  EXTREME SOLUTIONS (Best in Each Objective)")
    print(f"{'='*70}")
    
    if baseline_pareto and strategic_pareto:
        baseline_best_obj1 = max(baseline_pareto, key=lambda p: p[0])
        baseline_best_obj2 = max(baseline_pareto, key=lambda p: p[1])
        strategic_best_obj1 = max(strategic_pareto, key=lambda p: p[0])
        strategic_best_obj2 = max(strategic_pareto, key=lambda p: p[1])
        
        print(f"  Best Objective 1:")
        print(f"    Baseline:  {baseline_best_obj1}")
        print(f"    Strategic: {strategic_best_obj1}")
        if strategic_best_obj1[0] > baseline_best_obj1[0]:
            print(f"    Winner:    ✅ Strategic (+{strategic_best_obj1[0]-baseline_best_obj1[0]:.0f})")
        else:
            print(f"    Winner:    Baseline")
        
        print(f"\n  Best Objective 2:")
        print(f"    Baseline:  {baseline_best_obj2}")
        print(f"    Strategic: {strategic_best_obj2}")
        if strategic_best_obj2[1] > baseline_best_obj2[1]:
            print(f"    Winner:    ✅ Strategic (+{strategic_best_obj2[1]-baseline_best_obj2[1]:.0f})")
        else:
            print(f"    Winner:    Baseline")
        print()
    
    # 4. Statistical tests
    statistical_test(BASELINE_FILE, STRATEGIC_FILE, num_runs=10)
    
    # 5. Runtime breakdown (if stats file exists)
    if os.path.exists("llm_stats.csv"):
        print("5️⃣  RUNTIME BREAKDOWN")
        print(f"{'='*70}")
        # This would require storing runtime stats per run
        # For now, just placeholder
        print("  See main experiment output for runtime statistics\n")
    
    print(f"{'='*70}")
    print("✅ Extended analysis complete\n")


# Run extended analysis
# Uncomment after main experiment:
# extended_analysis()
```

```python
# ============================================================================
# EXPORT RESULTS FOR THESIS/PUBLICATION
# ============================================================================

def export_for_thesis():
    """
    Export results in LaTeX table format for thesis
    """
    print(f"\n{'='*70}")
    print("📄 EXPORTING RESULTS FOR THESIS")
    print(f"{'='*70}\n")
    
    BASELINE_FILE = "2502_ResultsMD.txt"
    STRATEGIC_FILE = "2502_Results_STRATEGIC.txt"
    
    baseline_all = load_results(BASELINE_FILE)
    strategic_all = load_results(STRATEGIC_FILE)
    
    baseline_pareto = get_true_pareto_front(baseline_all)
    strategic_pareto = get_true_pareto_front(strategic_all)
    
    # Calculate all metrics
    all_points = baseline_pareto + strategic_pareto
    if all_points:
        ref_point = [
            min(p[0] for p in all_points) - 100,
            min(p[1] for p in all_points) - 100
        ]
        
        hv_baseline = calculate_hypervolume(baseline_pareto, ref_point)
        hv_strategic = calculate_hypervolume(strategic_pareto, ref_point)
        
        spacing_baseline = calculate_spacing(baseline_pareto)
        spacing_strategic = calculate_spacing(strategic_pareto)
        
        # LaTeX table
        latex_table = f"""
% LaTeX table for thesis
\\begin{{table}}[htbp]
\\centering
\\caption{{Comparison of Baseline MOACP and Strategic LLM-MOACP on 250.2 instance}}
\\label{{tab:results_250_2}}
\\begin{{tabular}}{{lccr}}
\\toprule
\\textbf{{Metric}} & \\textbf{{Baseline}} & \\textbf{{Strategic LLM}} & \\textbf{{Improvement}} \\\\
\\midrule
Pareto Solutions    & {len(baseline_pareto)} & {len(strategic_pareto)} & {((len(strategic_pareto)-len(baseline_pareto))/len(baseline_pareto)*100):+.1f}\\% \\\\
Hypervolume         & {hv_baseline:,.0f} & {hv_strategic:,.0f} & {((hv_strategic-hv_baseline)/hv_baseline*100):+.1f}\\% \\\\
Spacing             & {spacing_baseline:.4f} & {spacing_strategic:.4f} & {((spacing_baseline-spacing_strategic)/spacing_baseline*100):+.1f}\\% \\\\
Runtime (10 runs)   & 10 min & - min & - \\\\
\\bottomrule
\\end{{tabular}}
\\end{{table}}
"""
        
        print("LaTeX Table:")
        print(latex_table)
        
        # Save to file
        with open("results_latex_table.tex", "w") as f:
            f.write(latex_table)
        print("✅ LaTeX table saved to: results_latex_table.tex\n")
        
        # CSV export
        csv_data = f"""Algorithm,Pareto_Size,Hypervolume,Spacing,Runtime_Minutes
Baseline,{len(baseline_pareto)},{hv_baseline:.2f},{spacing_baseline:.4f},10.0
Strategic_LLM,{len(strategic_pareto)},{hv_strategic:.2f},{spacing_strategic:.4f},-
"""
        
        with open("results_comparison.csv", "w") as f:
            f.write(csv_data)
        print("✅ CSV data saved to: results_comparison.csv\n")
        
        # Pareto fronts export
        with open("baseline_pareto.txt", "w") as f:
            for p in baseline_pareto:
                f.write(f"{p[0]:.6f} {p[1]:.6f}\n")
        
        with open("strategic_pareto.txt", "w") as f:
            for p in strategic_pareto:
                f.write(f"{p[0]:.6f} {p[1]:.6f}\n")
        
        print("✅ Pareto fronts saved to: baseline_pareto.txt, strategic_pareto.txt\n")
    
    print(f"{'='*70}")
    print("📁 Files ready for thesis:")
    print("  • results_latex_table.tex (ready to copy into thesis)")
    print("  • results_comparison.csv (for further analysis)")
    print("  • baseline_pareto.txt (Pareto front coordinates)")
    print("  • strategic_pareto.txt (Pareto front coordinates)")
    print(f"{'='*70}\n")


# Run export
# Uncomment after main experiment:
# export_for_thesis()
```
