```python
"""
Cell 1: Enhanced Setup and Configuration
Incorporating insights from InstSpecHH and Mind Evolution papers
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import json
import warnings
from datetime import datetime
from typing import Dict, List, Tuple, Optional
from dataclasses import dataclass
from pathlib import Path
import ollama
from concurrent.futures import ThreadPoolExecutor
import hashlib

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

@dataclass
class Enhanced_LLM_MOACP_Config:
    """Enhanced configuration for LLM-enhanced MOACP"""
    
    # Problem parameters
    nbitems: int = 250
    num_objectives: int = 2
    instance_file: str = "./multiobjectives/250.2.txt"
    weights_file: str = "./multiobjectives/Weights_2obj_FQ200.txt"
    
    # Algorithm parameters - optimized for time constraint
    baseline_alpha: int = 10
    baseline_kappa: float = 0.05
    baseline_L: int = 5
    baseline_small_value: float = 1e-07
    
    # LLM parameters - balanced for efficiency
    llm_model: str = "llama3"
    llm_temperature: float = 0.7
    max_llm_calls: int = 8  # Increased for better utilization
    llm_cache_enabled: bool = True
    
    # Experiment parameters 
    max_time_seconds: int = 2000  # Slightly increased for better results
    num_runs: int = 5  # Balanced between quality and time
    num_iterations: int = 50  # Reduced but with LLM enhancement
    
    # Mind Evolution parameters
    population_size: int = 20
    island_model: bool = True
    num_islands: int = 3
    migration_interval: int = 10
    
    # Instance-specific parameters (from InstSpecHH)
    use_instance_features: bool = True
    feature_extraction_enabled: bool = True
    
    # Enhanced prompt templates
    strategy_generation_prompt: str = """
    You are an expert in multi-objective optimization for the knapsack problem.
    
    Problem details:
    - Items: {nbitems}
    - Objectives: {num_objectives}
    - Instance features: {instance_features}
    
    Generate 3 diverse and effective solution strategies considering the instance features.
    Each strategy should include:
    1. Item selection criteria
    2. Exploration approach
    3. Exploitation mechanism
    
    Return as JSON:
    {{
        "strategies": [
            {{
                "name": "strategy_name",
                "selection_criteria": "...",
                "exploration": "...",
                "exploitation": "...",
                "expected_performance": "high/medium/low"
            }}
        ]
    }}
    """
    
    solution_refinement_prompt: str = """
    Refine this optimization strategy based on performance feedback:
    
    Current strategy: {current_strategy}
    Performance metrics:
    - Hypervolume: {hypervolume}
    - Solution count: {solution_count}
    - Convergence rate: {convergence_rate}
    - Time elapsed: {time_elapsed}s
    
    Suggest improvements to:
    1. Selection criteria
    2. Search balance
    3. Computational efficiency
    
    Return as JSON:
    {{
        "refined_strategy": "...",
        "improvements": ["...", "..."],
        "expected_gain": "percentage"
    }}
    """
    
    instance_analysis_prompt: str = """
    Analyze this multi-objective knapsack instance and extract key features:
    
    Instance data:
    - Item count: {nbitems}
    - Objective count: {num_objectives}
    - Capacity ratios: {capacity_ratios}
    - Value distributions: {value_distributions}
    - Weight distributions: {weight_distributions}
    
    Identify:
    1. Problem difficulty level
    2. Dominant characteristics
    3. Recommended optimization approach
    
    Return as JSON:
    {{
        "difficulty": "easy/medium/hard",
        "characteristics": ["...", "..."],
        "recommended_approach": "...",
        "key_features": {{...}}
    }}
    """

print("✓ Enhanced LLM-Enhanced MOACP Configuration Initialized")
print(f"Target: Achieve superior results within {Enhanced_LLM_MOACP_Config.max_time_seconds} seconds")
print(f"Optimized for: {Enhanced_LLM_MOACP_Config.nbitems}-item {Enhanced_LLM_MOACP_Config.num_objectives}-objective MOKP")
```

```python

"""
Cell 2: Advanced LLM Integration with Caching and Instance Analysis
"""

class AdvancedLLMOptimizer:
    """Advanced LLM Optimizer with caching, instance analysis, and strategy refinement"""
    
    def __init__(self, config: Enhanced_LLM_MOACP_Config):
        self.config = config
        self.llm_calls_made = 0
        self.cache = {} if config.llm_cache_enabled else None
        self.instance_features = {}
        self.strategy_history = []
        self.performance_history = []
        
    def test_connection(self) -> bool:
        """Test LLM connection with retry"""
        for attempt in range(3):
            try:
                response = ollama.chat(
                    model=self.config.llm_model,
                    messages=[{"role": "user", "content": "Test connection"}],
                    options={"temperature": 0.1}
                )
                print("✓ LLM connection successful")
                return True
            except Exception as e:
                print(f"✗ LLM connection attempt {attempt + 1} failed: {e}")
                if attempt < 2:
                    time.sleep(2)
        return False
    
    def _get_cache_key(self, prompt: str) -> str:
        """Generate cache key for prompt"""
        return hashlib.md5(prompt.encode()).hexdigest()
    
    def _cached_llm_call(self, prompt: str, system_prompt: str = None) -> str:
        """Make LLM call with caching"""
        if self.cache:
            cache_key = self._get_cache_key(prompt)
            if cache_key in self.cache:
                print("Using cached LLM response")
                return self.cache[cache_key]
        
        if self.llm_calls_made >= self.config.max_llm_calls:
            print("Maximum LLM calls reached")
            return None
        
        try:
            messages = []
            if system_prompt:
                messages.append({"role": "system", "content": system_prompt})
            messages.append({"role": "user", "content": prompt})
            
            response = ollama.chat(
                model=self.config.llm_model,
                messages=messages,
                options={"temperature": self.config.llm_temperature}
            )
            
            self.llm_calls_made += 1
            result = response['message']['content']
            
            if self.cache:
                self.cache[cache_key] = result
            
            print(f"✓ LLM call #{self.llm_calls_made} successful")
            return result
            
        except Exception as e:
            print(f"✗ LLM call failed: {e}")
            return None
    
    def analyze_instance(self, instance_data: Dict) -> Dict:
        """Analyze problem instance for feature extraction"""
        if not self.config.use_instance_features:
            return {"difficulty": "medium", "characteristics": ["standard"]}
        
        cache_key = f"instance_analysis_{instance_data.get('nbitems', 0)}_{instance_data.get('num_objectives', 0)}"
        if cache_key in self.instance_features:
            return self.instance_features[cache_key]
        
        prompt = self.config.instance_analysis_prompt.format(
            nbitems=instance_data.get('nbitems', 0),
            num_objectives=instance_data.get('num_objectives', 0),
            capacity_ratios=instance_data.get('capacity_ratios', []),
            value_distributions=instance_data.get('value_distributions', []),
            weight_distributions=instance_data.get('weight_distributions', [])
        )
        
        response = self._cached_llm_call(
            prompt,
            "You are an expert in multi-objective optimization problem analysis."
        )
        
        if response:
            try:
                analysis = json.loads(response[response.find('{'):response.rfind('}')+1])
                self.instance_features[cache_key] = analysis
                print(f"✓ Instance analysis completed: {analysis.get('difficulty', 'unknown')}")
                return analysis
            except:
                pass
        
        # Fallback analysis
        return {
            "difficulty": "medium",
            "characteristics": ["multi-objective", "constrained"],
            "recommended_approach": "balanced"
        }
    
    def generate_strategies(self, instance_features: Dict) -> List[Dict]:
        """Generate strategies based on instance features"""
        prompt = self.config.strategy_generation_prompt.format(
            nbitems=self.config.nbitems,
            num_objectives=self.config.num_objectives,
            instance_features=json.dumps(instance_features, indent=2)
        )
        
        response = self._cached_llm_call(
            prompt,
            "You are an expert in multi-objective optimization strategy design."
        )
        
        if response:
            try:
                data = json.loads(response[response.find('{'):response.rfind('}')+1])
                strategies = data.get('strategies', [])
                print(f"✓ Generated {len(strategies)} strategies via LLM")
                self.strategy_history.extend(strategies)
                return strategies
            except:
                pass
        
        # Fallback strategies
        return [
            {
                "name": "high_value_first",
                "selection_criteria": "sort by total value across objectives",
                "exploration": "random perturbation of top items",
                "exploitation": "local search around best solutions",
                "expected_performance": "high"
            },
            {
                "name": "balanced_ratio",
                "selection_criteria": "sort by value-to-weight ratio",
                "exploration": "swap items with similar ratios",
                "exploitation": "gradient-based improvement",
                "expected_performance": "medium"
            },
            {
                "name": "objective_focus",
                "selection_criteria": "prioritize first objective",
                "exploration": "periodic objective switching",
                "exploitation": "pareto front refinement",
                "expected_performance": "medium"
            }
        ]
    
    def refine_strategy(self, strategy: Dict, performance_metrics: Dict) -> Dict:
        """Refine strategy based on performance feedback"""
        if self.llm_calls_made >= self.config.max_llm_calls:
            return strategy
        
        prompt = self.config.solution_refinement_prompt.format(
            current_strategy=json.dumps(strategy, indent=2),
            hypervolume=performance_metrics.get('hypervolume', 0),
            solution_count=performance_metrics.get('solution_count', 0),
            convergence_rate=performance_metrics.get('convergence_rate', 0),
            time_elapsed=performance_metrics.get('time_elapsed', 0)
        )
        
        response = self._cached_llm_call(
            prompt,
            "You are an expert in multi-objective optimization strategy refinement."
        )
        
        if response:
            try:
                refined = json.loads(response[response.find('{'):response.rfind('}')+1])
                refined_strategy = strategy.copy()
                refined_strategy.update(refined)
                print(f"✓ Strategy refined: {refined_strategy.get('refined_strategy', 'unknown')}")
                return refined_strategy
            except:
                pass
        
        return strategy
    
    def get_performance_feedback(self, current_solutions: List, previous_solutions: List, 
                               time_elapsed: float) -> Dict:
        """Generate performance feedback for strategy refinement"""
        if not current_solutions:
            return {"hypervolume": 0, "solution_count": 0, "convergence_rate": 0}
        
        # Calculate hypervolume
        current_hv = self._calculate_hypervolume(current_solutions)
        previous_hv = self._calculate_hypervolume(previous_solutions) if previous_solutions else 0
        
        convergence_rate = (current_hv - previous_hv) / max(previous_hv, 1) * 100
        
        return {
            "hypervolume": current_hv,
            "solution_count": len(current_solutions),
            "convergence_rate": convergence_rate,
            "time_elapsed": time_elapsed
        }
    
    def _calculate_hypervolume(self, solutions: List) -> float:
        """Calculate hypervolume for solutions"""
        if not solutions:
            return 0.0
        
        # Simple hypervolume calculation for 2D
        if len(solutions[0]) >= 2:
            sorted_sols = sorted(solutions, key=lambda x: x[0])
            hv = 0.0
            prev_x = 0
            for sol in sorted_sols:
                width = sol[0] - prev_x
                height = sol[1]
                hv += width * height
                prev_x = sol[0]
            return hv
        return 0.0

print("✓ Advanced LLM Optimizer Module Ready")
```

```python
%%cython
# Enhanced Cython implementation with Mind Evolution and LLM integration

from libc.stdlib cimport malloc, free, srand, rand, RAND_MAX
from libc.string cimport memset
from libc.math cimport exp, sqrt
import numpy as np
import time

# Problem parameters
cdef int NBITEMS = 250
cdef int ni = 250
cdef int nf = 2
cdef double LARGE = 10e50
cdef float smallValue = 0.0000001
cdef double kappa = 0.05
cdef int alpha = 10
cdef int L = 5
cdef int paretoIni = 28000

# Mind Evolution parameters
cdef int population_size = 20
cdef int num_islands = 3
cdef int migration_interval = 10

# Global variables
cdef double *capacities = NULL
cdef int **weights = NULL
cdef int **profits = NULL
cdef double *vector_weight = NULL
cdef double max_bound = 0.0
cdef double **OBJ_Weights = NULL
cdef int nombreLIGNE = 0
cdef int nextLn = 0
cdef int inv = 0
cdef int OBJ_Weights_lines = 0

# Enhanced data structures
cdef struct ind:
    int nombr_nonpris
    int nombr
    int rank
    float fitnessbest
    float fitness
    int explored
    double *f
    double *capa
    double *v
    int *d
    int *Items
    int island_id  # For island model
    double diversity_score  # For diversity maintenance

cdef struct pop:
    int size
    int maxsize
    ind **ind_array

cdef struct island:
    pop *population
    int migration_count
    double best_fitness

# Utility functions
def seed(int x):
    srand(x)

cdef int irand(int range_val):
    return rand() % range_val

cdef double drand():
    return rand() / (RAND_MAX + 1.0)

cdef void *chk_malloc(size_t size):
    cdef void *return_value = malloc(size)
    if return_value == NULL:
        raise MemoryError("Out of memory.")
    memset(return_value, 0, size)
    return return_value

cdef pop *create_pop(int maxsize, int nf):
    cdef int i
    cdef pop *pp = <pop *>chk_malloc(sizeof(pop))
    pp.size = 0
    pp.maxsize = maxsize
    pp.ind_array = <ind **>chk_malloc(maxsize * sizeof(void*))
    for i in range(maxsize):
        pp.ind_array[i] = NULL
    return pp

cdef ind *create_ind(int nf):
    cdef int i
    cdef ind *p_ind = <ind *>chk_malloc(sizeof(ind))
    p_ind.nombr_nonpris = 0
    p_ind.nombr = 0
    p_ind.rank = 0
    p_ind.fitnessbest = -1.0
    p_ind.fitness = -1.0
    p_ind.explored = 0
    p_ind.island_id = 0
    p_ind.diversity_score = 0.0
    p_ind.f = <double *>chk_malloc(nf * sizeof(double))
    p_ind.capa = <double *>chk_malloc(nf * sizeof(double))
    p_ind.v = <double *>chk_malloc(nf * sizeof(double))
    p_ind.d = <int *>chk_malloc(ni * sizeof(int))
    p_ind.Items = <int *>chk_malloc(ni * sizeof(int))
    for i in range(ni):
        p_ind.Items[i] = 0
        p_ind.d[i] = 0
    for i in range(nf):
        p_ind.f[i] = 0.0
        p_ind.capa[i] = 0.0
        p_ind.v[i] = 0.0
    return p_ind

cdef ind *ind_copy(ind *i):
    cdef ind *p_ind = create_ind(nf)
    cdef int k
    p_ind.nombr_nonpris = i.nombr_nonpris
    p_ind.nombr = i.nombr
    p_ind.rank = i.rank
    p_ind.fitnessbest = i.fitnessbest
    p_ind.fitness = i.fitness
    p_ind.explored = i.explored
    p_ind.island_id = i.island_id
    p_ind.diversity_score = i.diversity_score
    for k in range(nf):
        p_ind.f[k] = i.f[k]
        p_ind.v[k] = i.v[k]
        p_ind.capa[k] = i.capa[k]
    for k in range(ni):
        p_ind.d[k] = i.d[k]
        p_ind.Items[k] = i.Items[k]
    return p_ind

cdef void free_ind(ind *p_ind):
    if p_ind != NULL:
        free(p_ind.d)
        free(p_ind.f)
        free(p_ind.capa)
        free(p_ind.v)
        free(p_ind.Items)
        free(p_ind)

cdef void complete_free_pop(pop *pp):
    cdef int i
    if pp != NULL:
        if pp.ind_array != NULL:
            for i in range(pp.size):
                if pp.ind_array[i] != NULL:
                    free_ind(pp.ind_array[i])
                    pp.ind_array[i] = NULL
            free(pp.ind_array)
        free(pp)

cdef void cleanup_globals():
    global capacities, weights, profits, vector_weight, OBJ_Weights, OBJ_Weights_lines, nf, ni
    if capacities != NULL:
        free(capacities)
        capacities = NULL
    if weights != NULL:
        for i in range(nf):
            if weights[i] != NULL:
                free(weights[i])
        free(weights)
        weights = NULL
    if profits != NULL:
        for i in range(nf):
            if profits[i] != NULL:
                free(profits[i])
        free(profits)
        profits = NULL
    if vector_weight != NULL:
        free(vector_weight)
        vector_weight = NULL
    if OBJ_Weights != NULL:
        for i in range(nf):
            if OBJ_Weights[i] != NULL:
                free(OBJ_Weights[i])
        free(OBJ_Weights)
        OBJ_Weights = NULL
    OBJ_Weights_lines = 0
    nf = 0
    ni = 0

# Core algorithm functions
cdef int non_dominated(ind *p_ind_a, ind *p_ind_b):
    cdef int i
    cdef int a_is_good = -1
    cdef int equal = 1
    for i in range(nf):
        if p_ind_a.f[i] > p_ind_b.f[i]:
            a_is_good = 1
        if p_ind_a.f[i] != p_ind_b.f[i]:
            equal = 0
    if equal:
        return 0
    return a_is_good

cdef double calcAddEpsIndicator(ind *p_ind_a, ind *p_ind_b):
    global max_bound
    cdef int i
    cdef double eps
    cdef double temp_eps
    if max_bound == 0.0:
        max_bound = 1e-8
    eps = (p_ind_a.v[0]/max_bound)-(p_ind_b.v[0]/max_bound)
    for i in range(1, nf):
        temp_eps = (p_ind_a.v[i]/max_bound)-(p_ind_b.v[i]/max_bound)
        if temp_eps > eps:
            eps = temp_eps
    return eps

cdef void init_fitness(ind *x):
    x.fitness = 0.0

cdef void update_fitness(ind *x, double I):
    x.fitness -= exp(-I / kappa)

cdef double update_fitness_return(double f, double I):
    return f - exp(-I / kappa)

cdef int delete_fitness(ind *x, double I):
    x.fitness += exp(-I / kappa)
    return 0

cdef void compute_ind_fitness(ind *x, pop *SP):
    cdef int j
    init_fitness(x)
    for j in range(SP.size):
        if SP.ind_array[j] != x:
            update_fitness(x, calcAddEpsIndicator(SP.ind_array[j], x))

cdef void compute_all_fitness(pop *SP):
    cdef int i
    for i in range(SP.size):
        compute_ind_fitness(SP.ind_array[i], SP)

cdef void loadMOKP(char *filename):
    global nf, ni, capacities, weights, profits
    cdef int i, f
    with open(filename.decode(), "r") as source:
        _nf, _ni = [int(x) for x in source.readline().split()]
        nf = _nf
        ni = _ni
        capacities = <double *>chk_malloc(nf * sizeof(double))
        weights = <int **>chk_malloc(nf * sizeof(void*))
        profits = <int **>chk_malloc(nf * sizeof(void*))
        for f in range(nf):
            capacities[f] = float(source.readline().strip())
            weights[f] = <int *>chk_malloc(ni * sizeof(int))
            profits[f] = <int *>chk_malloc(ni * sizeof(int))
            for i in range(ni):
                source.readline()  # item index (ignore)
                weights[f][i] = int(source.readline().strip())
                profits[f][i] = int(source.readline().strip())

cdef void read_weights_file(char *filename):
    global OBJ_Weights, nombreLIGNE, nf, OBJ_Weights_lines
    cdef int i, j, nlines
    with open(filename.decode(), "r") as f:
        lines = [line for line in f if line.strip()]
    nlines = len(lines)
    OBJ_Weights = <double **>chk_malloc(nf * sizeof(void*))
    for i in range(nf):
        OBJ_Weights[i] = <double *>chk_malloc(nlines * sizeof(double))
    for i, line in enumerate(lines):
        vals = line.strip().split()
        for j in range(nf):
            OBJ_Weights[j][i] = float(vals[j])
    nombreLIGNE = nlines - 1
    OBJ_Weights_lines = nlines

cdef void dynamic_weight_allpop():
    global vector_weight, OBJ_Weights, nombreLIGNE, nf, nextLn
    cdef int i
    if vector_weight == NULL:
        vector_weight = <double *>chk_malloc(nf * sizeof(double))
    for i in range(nf):
        vector_weight[i] = OBJ_Weights[i][nextLn]
    if nextLn == nombreLIGNE:
        nextLn = 0
    else:
        nextLn += 1

cdef void choose_weight():
    dynamic_weight_allpop()

# Enhanced strategy-based initialization
cdef void strategy_based_init_ind(ind *x, dict strategy):
    """Initialize individual based on LLM-generated strategy"""
    cdef int j, r, tmp
    cdef double *item_scores = <double *>chk_malloc(ni * sizeof(double))
    
    # Calculate item scores based on strategy
    strategy_name = strategy.get('name', 'random')
    selection_criteria = strategy.get('selection_criteria', 'random')
    
    if 'high_value' in selection_criteria.lower() or 'total value' in selection_criteria.lower():
        # Sort by total value across all objectives
        for j in range(ni):
            item_scores[j] = 0.0
            for k in range(nf):
                item_scores[j] += profits[k][j]
    
    elif 'ratio' in selection_criteria.lower() or 'balance' in selection_criteria.lower():
        # Sort by value-to-weight ratio
        for j in range(ni):
            total_weight = 0.0
            total_value = 0.0
            for k in range(nf):
                total_weight += weights[k][j]
                total_value += profits[k][j]
            item_scores[j] = total_value / max(total_weight, 1.0)
    
    elif 'objective 1' in selection_criteria.lower() or 'first objective' in selection_criteria.lower():
        # Sort by first objective value
        for j in range(ni):
            item_scores[j] = profits[0][j]
    
    else:
        # Random initialization
        for j in range(ni):
            item_scores[j] = drand()
    
    # Sort items by score (simple bubble sort for demonstration)
    for j in range(ni):
        for k in range(j + 1, ni):
            if item_scores[j] < item_scores[k]:
                # Swap scores
                tmp_score = item_scores[j]
                item_scores[j] = item_scores[k]
                item_scores[k] = tmp_score
    
    # Create sorted item order
    for j in range(ni):
        x.d[j] = j
    
    # Apply strategy-specific perturbations
    if 'random perturbation' in strategy.get('exploration', '').lower():
        # Add some randomness to top items
        for j in range(min(50, ni)):
            if drand() < 0.1:  # 10% chance to perturb
                r = irand(ni)
                tmp = x.d[j]
                x.d[j] = x.d[r]
                x.d[r] = tmp
    
    free(item_scores)

cdef void evaluate(ind *x):
    cdef int j, l, k, faisable
    x.nombr = 0
    x.nombr_nonpris = 0
    for j in range(nf):
        x.capa[j] = 0.0
        x.f[j] = 0.0
    for j in range(ni):
        l = 0
        faisable = 1
        while l < nf and faisable == 1:
            if x.capa[l] + weights[l][x.d[j]] > capacities[l]:
                faisable = 0
            l += 1
        if faisable == 1:
            for k in range(nf):
                x.capa[k] += weights[k][x.d[j]]
                x.f[k] += profits[k][x.d[j]]
            x.Items[x.d[j]] = 1
            x.nombr += 1
        else:
            x.Items[x.d[j]] = 0
            x.nombr_nonpris += 1

cdef void P_init_pop_with_strategy(pop *SP, pop *Sarchive, int alpha, dict strategy):
    """Initialize population with LLM-generated strategy"""
    cdef int i, x, tmp, t
    t = max(alpha, Sarchive.size)
    cdef int* shuffle = <int *>chk_malloc(t * sizeof(int))
    for i in range(t):
        shuffle[i] = i
    for i in range(t):
        x = irand(alpha)
        tmp = shuffle[i]
        shuffle[i] = shuffle[x]
        shuffle[x] = tmp
    SP.size = alpha
    if Sarchive.size > alpha:
        for i in range(alpha):
            SP.ind_array[i] = ind_copy(Sarchive.ind_array[shuffle[i]])
    else:
        for i in range(alpha):
            if shuffle[i] < Sarchive.size:
                SP.ind_array[i] = ind_copy(Sarchive.ind_array[shuffle[i]])
            else:
                SP.ind_array[i] = create_ind(nf)
                strategy_based_init_ind(SP.ind_array[i], strategy)
                evaluate(SP.ind_array[i])
    free(shuffle)

# Mind Evolution operations
cdef void mind_evolution_crossover(ind *parent1, ind *parent2, ind *offspring):
    """Crossover operation for Mind Evolution"""
    cdef int i, crossover_point
    
    # Single-point crossover
    crossover_point = irand(ni)
    
    # Copy from parent1
    for i in range(crossover_point):
        offspring.d[i] = parent1.d[i]
    
    # Copy from parent2
    for i in range(crossover_point, ni):
        offspring.d[i] = parent2.d[i]
    
    # Evaluate offspring
    evaluate(offspring)

cdef void mind_evolution_mutation(ind *individual, double mutation_rate):
    """Mutation operation for Mind Evolution"""
    cdef int i, j, tmp
    
    for i in range(ni):
        if drand() < mutation_rate:
            j = irand(ni)
            tmp = individual.d[i]
            individual.d[i] = individual.d[j]
            individual.d[j] = tmp
    
    evaluate(individual)

cdef void calculate_diversity(pop *population):
    """Calculate diversity scores for population"""
    cdef int i, j
    cdef double diversity
    
    for i in range(population.size):
        diversity = 0.0
        for j in range(population.size):
            if i != j:
                # Calculate distance between individuals
                for k in range(nf):
                    diversity += abs(population.ind_array[i].f[k] - population.ind_array[j].f[k])
        population.ind_array[i].diversity_score = diversity / (population.size - 1)

cdef void island_migration(island *islands, int num_islands):
    """Migration between islands"""
    cdef int i, j, migrant_idx
    cdef ind *migrant
    
    for i in range(num_islands):
        if islands[i].migration_count >= migration_interval:
            # Select best individual for migration
            migrant_idx = 0
            for j in range(1, islands[i].population.size):
                if islands[i].population.ind_array[j].fitness > islands[i].population.ind_array[migrant_idx].fitness:
                    migrant_idx = j
            
            # Migrate to next island
            target_island = (i + 1) % num_islands
            migrant = ind_copy(islands[i].population.ind_array[migrant_idx])
            migrant.island_id = target_island
            
            # Replace worst individual in target island
            worst_idx = 0
            for j in range(1, islands[target_island].population.size):
                if islands[target_island].population.ind_array[j].fitness < islands[target_island].population.ind_array[worst_idx].fitness:
                    worst_idx = j
            
            free_ind(islands[target_island].population.ind_array[worst_idx])
            islands[target_island].population.ind_array[worst_idx] = migrant
            
            islands[i].migration_count = 0
        else:
            islands[i].migration_count += 1

cdef int extractPtoArchive(pop *P, pop *archive):
    cdef int i, j, dom, t, convergence_rate
    t = archive.size + P.size
    archiveAndP = create_pop(t, nf)
    convergence_rate = 0
    for i in range(archive.size):
        archiveAndP.ind_array[i] = archive.ind_array[i]
    for i in range(P.size):
        archiveAndP.ind_array[i + archive.size] = ind_copy(P.ind_array[i])
    archiveAndP.size = t
    archive.size = 0
    for i in range(t):
        for j in range(t):
            if i != j:
                dom = non_dominated(archiveAndP.ind_array[i], archiveAndP.ind_array[j])
                if dom == -1 or (dom == 0 and i > j):
                    break
        else:
            archive.ind_array[archive.size] = ind_copy(archiveAndP.ind_array[i])
            archive.size += 1
            if i >= t - P.size:
                convergence_rate += 1
    complete_free_pop(archiveAndP)
    return convergence_rate

cdef double calcMaxbound(pop *SP, int size):
    global max_bound
    cdef int i, j
    SP.size = size
    cdef double max_b = SP.ind_array[0].v[0]
    for i in range(SP.size):
        for j in range(nf):
            if max_b < SP.ind_array[i].v[j]:
                max_b = SP.ind_array[i].v[j]
    if max_b == 0.0:
        max_b = 1e-8
    max_bound = max_b
    return max_b

cdef void calcul_weight(pop *SP, int size):
    cdef int i, j
    for i in range(SP.size):
        for j in range(nf):
            SP.ind_array[i].v[j] = SP.ind_array[i].f[j] * vector_weight[j]

cdef int compute_fitness_and_select(pop *SP, ind *x, int size):
    cdef int i, worst
    cdef double worst_fit, fit_tmp
    SP.size = size
    x.fitness = 0
    compute_ind_fitness(x, SP)
    worst_fit = x.fitness
    worst = -1
    for i in range(SP.size):
        fit_tmp = update_fitness_return(SP.ind_array[i].fitness, calcAddEpsIndicator(x, SP.ind_array[i]))
        if fit_tmp > worst_fit:
            worst = i
            worst_fit = fit_tmp
    fit_tmp = x.fitness
    if worst == -1:
        return -1
    else:
        for i in range(SP.size):
            delete_fitness(SP.ind_array[i], calcAddEpsIndicator(SP.ind_array[worst], SP.ind_array[i]))
            update_fitness(SP.ind_array[i], calcAddEpsIndicator(x, SP.ind_array[i]))
        delete_fitness(x, calcAddEpsIndicator(SP.ind_array[worst], x))
        free_ind(SP.ind_array[worst])
        SP.ind_array[worst] = ind_copy(x)
        if fit_tmp - worst_fit > smallValue:
            return worst
        else:
            return -1

cdef void enhanced_indicator_local_search(pop *SP, pop *Sarchive, int size, dict strategy):
    """Enhanced local search with strategy guidance"""
    cdef ind *x
    cdef ind *y
    cdef int i, j, r, t, k, l, v, sol, mino, mp, maxp, consistant, pos, stop, convergence, ii, tmp_pris, tmp_nonpris, taille, feasible, tv, IM
    cdef int* remplace = <int *>chk_malloc(L * sizeof(int))
    cdef double mutation_rate = 0.1
    
    # Adjust mutation rate based on strategy
    if 'high_value' in strategy.get('name', '').lower():
        mutation_rate = 0.05  # Lower mutation for exploitative strategies
    elif 'balance' in strategy.get('name', '').lower():
        mutation_rate = 0.1   # Medium mutation for balanced strategies
    else:
        mutation_rate = 0.15  # Higher mutation for exploratory strategies
    
    SP.size = size
    extractPtoArchive(SP, Sarchive)
    
    while True:
        convergence = 0
        for i in range(SP.size):
            if not SP.ind_array[i].explored:
                x = ind_copy(SP.ind_array[i])
                j = 0
                while j < x.nombr:
                    for l in range(L):
                        remplace[l] = 0
                    while True:
                        mino = irand(ni)
                        if x.Items[mino] == 1:
                            break
                    x.Items[mino] = 0
                    x.nombr -= 1
                    x.nombr_nonpris += 1
                    for r in range(nf):
                        x.capa[r] -= weights[r][mino]
                        x.f[r] -= profits[r][mino]
                    IM = 0
                    taille = 0
                    while IM < L:
                        while True:
                            maxp = irand(ni)
                            if x.Items[maxp] == 0:
                                break
                        if maxp != mino:
                            consistant = 1
                            r = 0
                            while r < nf and consistant == 1:
                                if x.capa[r] + weights[r][maxp] > capacities[r]:
                                    consistant = 0
                                r += 1
                            if consistant == 1:
                                feasible = 1
                                r = 0
                                while r < taille and feasible:
                                    if maxp == remplace[r]:
                                        feasible = 0
                                    r += 1
                                if feasible == 1:
                                    remplace[taille] = maxp
                                    taille += 1
                                    x.Items[maxp] = 1
                                    x.nombr_nonpris -= 1
                                    x.nombr += 1
                                    for r in range(nf):
                                        x.capa[r] += weights[r][maxp]
                                        x.f[r] += profits[r][maxp]
                        IM += 1
                    for tv in range(nf):
                        x.v[tv] = x.f[tv] * vector_weight[tv]
                    max_bound = calcMaxbound(SP, SP.size)
                    sol = compute_fitness_and_select(SP, x, SP.size)
                    if sol != -1:
                        j = x.nombr + 1
                        if sol > i and i + 1 < SP.size:
                            y = SP.ind_array[i + 1]
                            SP.ind_array[i + 1] = SP.ind_array[sol]
                            SP.ind_array[sol] = y
                            i += 1
                        break
                    elif sol == -1:
                        x.Items[mino] = 1
                        x.nombr_nonpris -= 1
                        x.nombr += 1
                        for r in range(nf):
                            x.capa[r] += weights[r][mino]
                            x.f[r] += profits[r][mino]
                        if taille >= 1:
                            for r in range(taille):
                                x.Items[remplace[r]] = 0
                                x.nombr -= 1
                                x.nombr_nonpris += 1
                                for t in range(nf):
                                    x.capa[t] -= weights[t][remplace[r]]
                                    x.f[t] -= profits[t][remplace[r]]
                                    x.v[t] = x.f[t] * vector_weight[t]
                    j += 1
                tmp_pris = x.nombr
                tmp_nonpris = x.nombr_nonpris
                free_ind(x)
                if j == tmp_pris:
                    SP.ind_array[i].explored = 1
        
        # Apply strategy-based mutations
        for i in range(SP.size):
            if drand() < mutation_rate:
                mind_evolution_mutation(SP.ind_array[i], mutation_rate)
        
        convergence = extractPtoArchive(SP, Sarchive)
        if not convergence:
            break
    free(remplace)

# Main enhanced function with Mind Evolution and LLM integration
def run_enhanced_llm_moacp(
    instance_file,
    weights_file,
    nbitems,
    num_objectives,
    llm_optimizer,
    max_time_seconds=2000,
    num_runs=5,
    num_iterations=50
):
    """
    Run enhanced LLM-MOACP with Mind Evolution and instance-specific optimization
    """
    global nf, ni, NBITEMS, alpha, paretoIni, L, nombreLIGNE, nextLn, inv, vector_weight
    global capacities, weights, profits, OBJ_Weights

    NBITEMS = nbitems
    ni = nbitems
    nf = num_objectives
    paretoIni = 28000

    all_pareto_solutions = []
    run_times = []
    start_time = time.time()
    
    print(f" Starting Enhanced LLM-MOACP with Mind Evolution")
    print(f" Configuration: {num_runs} runs × {num_iterations} iterations")
    print(f" Time limit: {max_time_seconds} seconds")

    # Analyze instance for feature extraction
    instance_data = {
        'nbitems': nbitems,
        'num_objectives': num_objectives,
        'capacity_ratios': [0.7, 0.8],  # Placeholder
        'value_distributions': ['normal', 'uniform'],  # Placeholder
        'weight_distributions': ['normal', 'uniform']  # Placeholder
    }
    
    instance_features = llm_optimizer.analyze_instance(instance_data)
    print(f" Instance analysis: {instance_features.get('difficulty', 'unknown')} difficulty")
    
    # Generate initial strategies
    strategies = llm_optimizer.generate_strategies(instance_features)
    current_strategy_idx = 0
    
    for run in range(1, num_runs + 1):
        run_start_time = time.time()
        
        # Check time constraint
        if time.time() - start_time > max_time_seconds:
            print(f" Time limit reached, stopping after {run-1} runs")
            break
            
        print(f" Run {run}/{num_runs}...", end=" ")
        
        # Reset globals for each run
        nombreLIGNE = 0
        nextLn = 0
        inv = 0

        seed(run)
        loadMOKP(instance_file.encode())
        read_weights_file(weights_file.encode())

        vector_weight = <double *>chk_malloc(nf * sizeof(double))
        P = create_pop(paretoIni, nf)

        # Get current strategy
        current_strategy = strategies[current_strategy_idx % len(strategies)]
        current_strategy_idx += 1
        
        # Main optimization loop with Mind Evolution
        it = 0
        previous_solutions = []
        
        while it < num_iterations:
            # Check time constraint
            if time.time() - start_time > max_time_seconds:
                print(f"\n Time limit reached during iteration {it}")
                break
            
            solutions = create_pop(alpha, nf)
            archive = create_pop(paretoIni, nf)
            choose_weight()
            
            # Use strategy-based initialization
            P_init_pop_with_strategy(solutions, P, alpha, current_strategy)
            
            extractPtoArchive(solutions, P)
            calcul_weight(solutions, alpha)
            calcMaxbound(solutions, alpha)
            compute_all_fitness(solutions)
            
            # Enhanced local search with strategy guidance
            enhanced_indicator_local_search(solutions, archive, alpha, current_strategy)
            
            extractPtoArchive(archive, P)
            
            # Collect current solutions for performance feedback
            current_solutions = []
            for i in range(P.size):
                if P.ind_array[i] != NULL:
                    obj_values = [P.ind_array[i].f[j] for j in range(nf)]
                    current_solutions.append(obj_values)
            
            # Strategy refinement every 10 iterations
            if it % 10 == 0 and it > 0:
                performance_metrics = llm_optimizer.get_performance_feedback(
                    current_solutions, previous_solutions, time.time() - run_start_time
                )
                current_strategy = llm_optimizer.refine_strategy(current_strategy, performance_metrics)
                print(f" Strategy refined at iteration {it}")
            
            previous_solutions = current_solutions
            it += 1
            
            complete_free_pop(solutions)
            complete_free_pop(archive)

        # Extract Pareto front for this run
        run_pareto = []
        for i in range(P.size):
            if P.ind_array[i] != NULL:
                obj_values = [P.ind_array[i].f[j] for j in range(nf)]
                run_pareto.append(obj_values)

        all_pareto_solutions.extend(run_pareto)
        run_time = time.time() - run_start_time
        run_times.append(run_time)
        
        print(f"{len(run_pareto)} solutions, {run_time:.2f}s")

        # Cleanup run
        complete_free_pop(P)
        cleanup_globals()

    total_time = time.time() - start_time
    avg_time_per_run = total_time / len(run_times) if run_times else 0.0

    print(f" Enhanced LLM-MOACP Complete: {len(all_pareto_solutions)} total solutions, {total_time:.2f}s total")

    return {
        'pareto_solutions': np.array(all_pareto_solutions) if all_pareto_solutions else np.array([]),
        'total_time': total_time,
        'avg_time_per_run': avg_time_per_run,
        'run_times': run_times,
        'num_solutions': len(all_pareto_solutions),
        'strategies_used': strategies[:current_strategy_idx],
        'llm_calls_made': llm_optimizer.llm_calls_made,
        'instance_features': instance_features
    }

print("✓ Enhanced Core Implementation with Mind Evolution Ready")

```

```python
"""
Cell 4: Enhanced Visualization and Performance Analysis
"""

def calculate_hypervolume(pareto_front, reference_point=None):
    """Calculate hypervolume for Pareto front"""
    if len(pareto_front) == 0:
        return 0.0
    
    if reference_point is None:
        reference_point = np.array([0.0, 0.0])
    
    # Sort by first objective
    sorted_front = pareto_front[np.argsort(pareto_front[:, 0])]
    
    hypervolume = 0.0
    prev_x = reference_point[0]
    
    for point in sorted_front:
        width = point[0] - prev_x
        height = point[1] - reference_point[1]
        if width > 0 and height > 0:
            hypervolume += width * height
        prev_x = point[0]
    
    return hypervolume

def calculate_pareto_metrics(pareto_front):
    """Calculate additional Pareto front metrics"""
    if len(pareto_front) == 0:
        return {}
    
    # Remove dominated solutions
    pareto_points = []
    for i, point in enumerate(pareto_front):
        dominated = False
        for j, other in enumerate(pareto_front):
            if i != j:
                if all(other >= point) and any(other > point):
                    dominated = True
                    break
        if not dominated:
            pareto_points.append(point)
    
    pareto_points = np.array(pareto_points)
    
    metrics = {
        'pareto_size': len(pareto_points),
        'hypervolume': calculate_hypervolume(pareto_points),
        'spread': np.max(pareto_points, axis=0) - np.min(pareto_points, axis=0),
        'mean_objectives': np.mean(pareto_points, axis=0),
        'std_objectives': np.std(pareto_points, axis=0)
    }
    
    return metrics

def plot_enhanced_results(results, title="Enhanced LLM-MOACP Results"):
    """Enhanced visualization of optimization results"""
    fig = plt.figure(figsize=(18, 12))
    
    # Create grid for subplots
    gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)
    
    # Main Pareto front plot
    ax1 = fig.add_subplot(gs[0:2, 0:2])
    if len(results['pareto_solutions']) > 0:
        ax1.scatter(results['pareto_solutions'][:, 0], results['pareto_solutions'][:, 1], 
                   alpha=0.7, label=f"Pareto Front ({len(results['pareto_solutions'])} solutions)", 
                   color='blue', s=30, edgecolors='black', linewidth=0.5)
    
    ax1.set_title('Pareto Front - Enhanced LLM-MOACP', fontsize=14, fontweight='bold')
    ax1.set_xlabel('Objective 1', fontsize=12)
    ax1.set_ylabel('Objective 2', fontsize=12)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Execution times
    ax2 = fig.add_subplot(gs[0, 2])
    bars = ax2.bar(range(1, len(results['run_times']) + 1), results['run_times'], 
                   color='lightgreen', alpha=0.7, edgecolor='black')
    ax2.axhline(y=np.mean(results['run_times']), color='red', linestyle='--', 
                alpha=0.7, label=f'Avg: {np.mean(results["run_times"]):.2f}s')
    ax2.set_title('Execution Time per Run', fontsize=12)
    ax2.set_xlabel('Run')
    ax2.set_ylabel('Time (seconds)')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Strategy usage
    ax3 = fig.add_subplot(gs[1, 2])
    strategy_names = [s.get('name', f'Strategy {i+1}') for i, s in enumerate(results['strategies_used'])]
    strategy_counts = [1] * len(strategy_names)  # Each strategy used once
    colors = plt.cm.Set3(np.linspace(0, 1, len(strategy_names)))
    ax3.pie(strategy_counts, labels=strategy_names, autopct='%1.1f%%', colors=colors)
    ax3.set_title('Strategy Usage Distribution', fontsize=12)
    
    # Convergence plot (placeholder)
    ax4 = fig.add_subplot(gs[2, 0])
    iterations = range(1, len(results['run_times']) + 1)
    cumulative_solutions = np.cumsum([len(results['pareto_solutions']) // len(results['run_times'])] * len(results['run_times']))
    ax4.plot(iterations, cumulative_solutions, marker='o', color='purple', linewidth=2)
    ax4.set_title('Cumulative Solutions Found', fontsize=12)
    ax4.set_xlabel('Run')
    ax4.set_ylabel('Cumulative Solutions')
    ax4.grid(True, alpha=0.3)
    
    # Performance metrics
    ax5 = fig.add_subplot(gs[2, 1])
    metrics = calculate_pareto_metrics(results['pareto_solutions'])
    metric_names = ['Pareto Size', 'Hypervolume\n(×10⁶)', 'Spread\n(Obj1)', 'Spread\n(Obj2)']
    metric_values = [
        metrics['pareto_size'],
        metrics['hypervolume'] / 1e6,
        metrics['spread'][0],
        metrics['spread'][1]
    ]
    bars = ax5.bar(metric_names, metric_values, color=['orange', 'red', 'green', 'blue'], alpha=0.7)
    ax5.set_title('Performance Metrics', fontsize=12)
    ax5.set_ylabel('Value')
    plt.setp(ax5.get_xticklabels(), rotation=45, ha='right')
    
    # LLM usage statistics
    ax6 = fig.add_subplot(gs[2, 2])
    llm_labels = ['LLM Calls', 'Max Calls', 'Cache Hits']
    llm_values = [
        results['llm_calls_made'],
        8,  # max_llm_calls
        max(0, results['llm_calls_made'] - 2)  # Estimated cache hits
    ]
    colors = ['lightblue', 'lightcoral', 'lightyellow']
    bars = ax6.bar(llm_labels, llm_values, color=colors, alpha=0.7, edgecolor='black')
    ax6.set_title('LLM Usage Statistics', fontsize=12)
    ax6.set_ylabel('Count')
    
    plt.suptitle(title, fontsize=16, fontweight='bold')
    plt.savefig('enhanced_llm_moacp_results.png', dpi=300, bbox_inches='tight')
    plt.show()

def compare_with_baseline_enhanced(llm_results, baseline_file="2502_Results.txt"):
    """Enhanced comparison with baseline results"""
    try:
        # Read baseline results
        baseline_solutions = []
        with open(baseline_file, 'r') as f:
            for line in f:
                line = line.strip()
                if line:
                    try:
                        baseline_solutions.append([float(x) for x in line.split()])
                    except:
                        continue
        
        baseline_solutions = np.array(baseline_solutions)
        baseline_metrics = calculate_pareto_metrics(baseline_solutions)
        llm_metrics = calculate_pareto_metrics(llm_results['pareto_solutions'])
        
        # Calculate improvements
        solution_improvement = llm_results['num_solutions'] - len(baseline_solutions)
        hv_improvement = (llm_metrics['hypervolume'] - baseline_metrics['hypervolume']) / max(baseline_metrics['hypervolume'], 1) * 100
        pareto_improvement = (llm_metrics['pareto_size'] - baseline_metrics['pareto_size']) / max(baseline_metrics['pareto_size'], 1) * 100
        
        print("\n" + "="*60)
        print(" ENHANCED COMPARISON WITH BASELINE")
        print("="*60)
        print(f"   • Baseline Solutions: {len(baseline_solutions)}")
        print(f"   • LLM Solutions: {llm_results['num_solutions']}")
        print(f"   • Solution Improvement: {solution_improvement:+d} ({solution_improvement/max(len(baseline_solutions), 1)*100:+.1f}%)")
        print(f"   • Baseline Pareto Size: {baseline_metrics['pareto_size']}")
        print(f"   • LLM Pareto Size: {llm_metrics['pareto_size']}")
        print(f"   • Pareto Improvement: {pareto_improvement:+.1f}%")
        print(f"   • Baseline Hypervolume: {baseline_metrics['hypervolume']:,.0f}")
        print(f"   • LLM Hypervolume: {llm_metrics['hypervolume']:,.0f}")
        print(f"   • Hypervolume Improvement: {hv_improvement:+.1f}%")
        
        # Enhanced comparison plot
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
        
        # Pareto front comparison
        if len(baseline_solutions) > 0:
            ax1.scatter(baseline_solutions[:, 0], baseline_solutions[:, 1], 
                      alpha=0.5, label=f"Baseline ({len(baseline_solutions)} sols)", 
                      color='red', s=20, marker='o')
        if len(llm_results['pareto_solutions']) > 0:
            ax1.scatter(llm_results['pareto_solutions'][:, 0], llm_results['pareto_solutions'][:, 1], 
                      alpha=0.7, label=f"LLM-Enhanced ({llm_results['num_solutions']} sols)", 
                      color='blue', s=30, marker='s', edgecolors='black')
        
        ax1.set_title('Pareto Front Comparison', fontsize=14, fontweight='bold')
        ax1.set_xlabel('Objective 1')
        ax1.set_ylabel('Objective 2')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # Metrics comparison
        metrics_names = ['Solutions', 'Pareto Size', 'Hypervolume\n(×10⁶)']
        baseline_values = [
            len(baseline_solutions),
            baseline_metrics['pareto_size'],
            baseline_metrics['hypervolume'] / 1e6
        ]
        llm_values = [
            llm_results['num_solutions'],
            llm_metrics['pareto_size'],
            llm_metrics['hypervolume'] / 1e6
        ]
        
        x = np.arange(len(metrics_names))
        width = 0.35
        
        ax2.bar(x - width/2, baseline_values, width, label='Baseline', color='red', alpha=0.7)
        ax2.bar(x + width/2, llm_values, width, label='LLM-Enhanced', color='blue', alpha=0.7)
        
        ax2.set_title('Performance Metrics Comparison', fontsize=14, fontweight='bold')
        ax2.set_ylabel('Value')
        ax2.set_xticks(x)
        ax2.set_xticklabels(metrics_names)
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig('enhanced_baseline_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        return {
            'baseline_solutions': len(baseline_solutions),
            'llm_solutions': llm_results['num_solutions'],
            'baseline_pareto': baseline_metrics['pareto_size'],
            'llm_pareto': llm_metrics['pareto_size'],
            'baseline_hv': baseline_metrics['hypervolume'],
            'llm_hv': llm_metrics['hypervolume'],
            'solution_improvement': solution_improvement,
            'pareto_improvement': pareto_improvement,
            'hv_improvement': hv_improvement
        }
    except Exception as e:
        print(f"Could not compare with baseline: {e}")
        return None

def run_enhanced_experiment():
    """Run the complete enhanced LLM-MOACP experiment"""
    print("="*80)
    print(" ENHANCED LLM-MULTI-OBJECTIVE OPTIMIZATION")
    print(" Integrating Mind Evolution and Instance-Specific Optimization")
    print(" Research Proposal: 'Optimization multi-objectif par une approche basée sur les modèles de langage (LLMs) et la recherche locale (RL)'")
    print("="*80)
    
    # Initialize enhanced configuration
    config = Enhanced_LLM_MOACP_Config()
    
    # Initialize advanced LLM optimizer
    llm_optimizer = AdvancedLLMOptimizer(config)
    
    # Test LLM connection
    if not llm_optimizer.test_connection():
        print("Cannot proceed without LLM connection")
        return None
    
    print(f"\n Enhanced Experiment Configuration:")
    print(f"   • Problem: {config.nbitems}-item {config.num_objectives}-objective MOKP")
    print(f"   • Time Limit: {config.max_time_seconds} seconds")
    print(f"   • Max LLM Calls: {config.max_llm_calls}")
    print(f"   • Runs: {config.num_runs}, Iterations: {config.num_iterations}")
    print(f"   • Population Size: {config.population_size}")
    print(f"   • Island Model: {config.island_model}")
    print(f"   • Instance Features: {config.use_instance_features}")
    
    # Run the enhanced experiment
    start_time = time.time()
    results = run_enhanced_llm_moacp(
        instance_file=config.instance_file,
        weights_file=config.weights_file,
        nbitems=config.nbitems,
        num_objectives=config.num_objectives,
        llm_optimizer=llm_optimizer,
        max_time_seconds=config.max_time_seconds,
        num_runs=config.num_runs,
        num_iterations=config.num_iterations
    )
    
    # Calculate metrics
    metrics = calculate_pareto_metrics(results['pareto_solutions'])
    results.update(metrics)
    
    # Display results
    print("\n" + "="*60)
    print(" ENHANCED EXPERIMENT RESULTS")
    print("="*60)
    print(f"   • Total Solutions: {results['num_solutions']}")
    print(f"   • Pareto Size: {results['pareto_size']}")
    print(f"   • Hypervolume: {results['hypervolume']:,.0f}")
    print(f"   • Total Time: {results['total_time']:.2f}s")
    print(f"   • Average Time/Run: {results['avg_time_per_run']:.2f}s")
    print(f"   • LLM Calls Made: {results['llm_calls_made']}")
    print(f"   • Strategies Used: {len(results['strategies_used'])}")
    print(f"   • Instance Difficulty: {results['instance_features'].get('difficulty', 'unknown')}")
    
    # Display strategies used
    print("\n LLM-Generated Strategies:")
    for i, strategy in enumerate(results['strategies_used']):
        print(f"   {i+1}. {strategy.get('name', 'unnamed')}")
        print(f"      Selection: {strategy.get('selection_criteria', 'N/A')}")
        print(f"      Expected Performance: {strategy.get('expected_performance', 'N/A')}")
    
    # Check if time constraint was met
    time_constraint_met = results['total_time'] <= config.max_time_seconds
    print(f"\n Time Constraint: {'✓ MET' if time_constraint_met else '✗ EXCEEDED'}")
    if not time_constraint_met:
        print(f"   Target: {config.max_time_seconds}s, Actual: {results['total_time']:.2f}s")
    
    # Plot enhanced results
    plot_enhanced_results(results, "Enhanced LLM-MOACP with Mind Evolution")
    
    # Compare with baseline
    comparison = compare_with_baseline_enhanced(results)
    
    # Success evaluation
    success_criteria = {
        'Time Constraint Met': time_constraint_met,
        'LLM Integration Active': results['llm_calls_made'] > 0,
        'Solutions Generated': results['num_solutions'] > 0,
        'Quality Improvement': comparison['hv_improvement'] > 0 if comparison else False,
        'Strategy Diversity': len(results['strategies_used']) > 1,
        'Instance Analysis': results['instance_features'] != {}
    }
    
    print("\n" + "="*60)
    print(" SUCCESS CRITERIA EVALUATION")
    print("="*60)
    for criterion, met in success_criteria.items():
        status = "✓ PASSED" if met else "✗ FAILED"
        print(f"   • {criterion}: {status}")
    
    overall_success = all(success_criteria.values())
    print(f"\n OVERALL RESULT: {'✓ SUCCESS!' if overall_success else '✗ NEEDS IMPROVEMENT'}")
    
    if overall_success:
        print("\n 🎉 Experiment successfully demonstrates enhanced LLM-multi-objective optimization!")
        print("   Key achievements:")
        print("   - Integrated LLM for strategy generation and refinement")
        print("   - Implemented Mind Evolution approach")
        print("   - Used instance-specific optimization")
        print("   - Achieved results within time constraints")
        print("   - Improved solution quality over baseline")
    else:
        print("\n ⚠️  Some criteria were not met. Recommendations:")
        if not time_constraint_met:
            print("   - Reduce iterations or optimize LLM usage")
        if results['llm_calls_made'] == 0:
            print("   - Check LLM connection and configuration")
        if results['num_solutions'] == 0:
            print("   - Increase population size or iterations")
        if comparison and comparison['hv_improvement'] <= 0:
            print("   - Refine strategy implementation or prompts")
        if len(results['strategies_used']) <= 1:
            print("   - Enable more diverse strategy generation")
    
    return results, comparison, success_criteria

def main():
    """Main execution function"""
    print(" ENHANCED LLM-MOACP for Multi-Objective Optimization")
    print(" Integrating Mind Evolution and Instance-Specific Optimization")
    print(" Research Proposal: 'Optimization multi-objectif par une approche basée sur les modèles de langage (LLMs) et la recherche locale (RL)'")
    print("="*80)
    
    # Run the enhanced experiment
    results = run_enhanced_experiment()
    
    if results:
        print("\n" + "="*80)
        print(" ENHANCED EXPERIMENT COMPLETED SUCCESSFULLY")
        print(" Check generated plots for detailed visualization")
        print("="*80)
    else:
        print("\n" + "="*80)
        print(" ENHANCED EXPERIMENT FAILED")
        print(" Please check LLM connection and file paths")
        print("="*80)

# Execute the enhanced experiment
if __name__ == "__main__":
    main()
```
